## What you will learn
Cell Painting is a high-content imaging assay that captures how a drug changes
cell morphology. By measuring thousands of features from fluorescence microscopy images,
we can infer mechanism of action without knowing anything about the target in advance.

### Pipeline (Part A + B merged)
1. **Image features:** What Cell Painting measures, how CellProfiler extracts morphological features
2. **Feature processing:** Normalisation, feature selection, dimensionality reduction
3. **Clustering:** Grouping compounds by morphological similarity → mechanism of action clusters
4. **Similarity search:** Finding compounds with the same morphological profile as a known drug
5. **Visualisation:** UMAP of morphological profiles, annotated with known mechanisms

## Why Cell Painting matters
- Captures drug effects on ALL cellular processes simultaneously
- Works even for compounds with unknown targets
- One assay generates ~1000+ features (much richer than IC50 alone)
- Used by Broad Institute, Recursion Pharmaceuticals, and others at scale

**Data source:** JUMP-CP dataset (public) or simulated data for this demo

```python
# ============================================================
# CELL 1: Install libraries
# ============================================================
print('Installing packages...')
import subprocess, sys
subprocess.run([sys.executable, '-m', 'pip', 'install', '--quiet',
                'pandas', 'numpy', 'matplotlib', 'scikit-learn', 'umap-learn',
                'seaborn'], check=False)
print('Done!')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import umap

print(f'NumPy   : {np.__version__}')
print(f'Pandas  : {pd.__version__}')
print('Libraries imported!')

# Colour palette for mechanism clusters
MOA_COLORS = {
    'EGFR_inhibitor':    '#0D9488',
    'microtubule':       '#E57373',
    'DNA_damage':        '#F59E0B',
    'proteasome':        '#6B7280',
    'actin_cytoskeleton':'#1B2A4A',
    'unknown':           '#D1D5DB',
}
```

```python
# ============================================================
# CELL 2: Simulate a Cell Painting dataset
# ============================================================
# Real Cell Painting data comes from CellProfiler image analysis.
# Features are grouped into 5 channels × multiple feature types:
#   DNA (Hoechst), ER (concanavalin A), RNA (SYTO14),
#   AGP (phalloidin+WGA), Mito (MitoTracker)
# Typical feature set: ~1000-3000 features per cell, then aggregated per well.

print('=== Simulating Cell Painting Morphological Profiles ===')
print('In practice, these come from CellProfiler image analysis of microscopy images.')
print('Features cover: nucleus shape, ER texture, actin organisation, mitochondria, etc.\n')

np.random.seed(42)
N_COMPOUNDS  = 150   # number of compounds
N_FEATURES   = 200   # morphological features per compound

# Define mechanism of action (MoA) groups with distinct morphological signatures
MOA_GROUPS = {
    'EGFR_inhibitor':     (np.array([2.0, -1.0, 0.5, 1.5] * 50), 0.5),   # centroid, noise
    'microtubule':        (np.array([-2.0, 2.0, -0.5, 0.5] * 50), 0.6),
    'DNA_damage':         (np.array([1.0, 1.5, -2.0, -1.0] * 50), 0.7),
    'proteasome':         (np.array([-1.0, -1.5, 1.0, 2.0] * 50), 0.6),
    'actin_cytoskeleton': (np.array([0.5, -2.0, 2.0, -0.5] * 50), 0.5),
}
N_PER_GROUP  = N_COMPOUNDS // len(MOA_GROUPS)

profiles, labels, compound_names = [], [], []
for moa, (centroid, noise_sd) in MOA_GROUPS.items():
    for i in range(N_PER_GROUP):
        profile = centroid[:N_FEATURES] + np.random.normal(0, noise_sd, N_FEATURES)
        profiles.append(profile)
        labels.append(moa)
        compound_names.append(f'{moa[:4].upper()}-{i+1:03d}')

# Remaining compounds are 'unknown'
for i in range(N_COMPOUNDS - len(profiles)):
    profiles.append(np.random.normal(0, 1.0, N_FEATURES))
    labels.append('unknown')
    compound_names.append(f'UNK-{i+1:03d}')

df_profiles = pd.DataFrame(
    np.array(profiles),
    columns=[f'feat_{i:03d}' for i in range(N_FEATURES)]
)
df_profiles['compound'] = compound_names
df_profiles['true_moa'] = labels

print(f'Dataset: {len(df_profiles)} compounds × {N_FEATURES} morphological features')
print('MoA distribution:')
for moa, n in df_profiles['true_moa'].value_counts().items():
    print(f'  {moa:<25}: {n} compounds')
```

```python
# ============================================================
# CELL 3: Feature normalisation and PCA
# ============================================================
# Cell Painting features have very different ranges and scales.
# StandardScaler: subtract mean, divide by std → mean=0, std=1 for each feature.
# Then PCA reduces 200 features to 50 principal components (captures 90%+ variance).

FEAT_COLS = [c for c in df_profiles.columns if c.startswith('feat_')]
X = df_profiles[FEAT_COLS].values.astype(np.float32)

print('Step 1: Normalising features (zero mean, unit variance)...')
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(f'  Before scaling: mean={X.mean():.3f}  std={X.std():.3f}')
print(f'  After  scaling: mean={X_scaled.mean():.3f}  std={X_scaled.std():.3f}')

print('\nStep 2: PCA dimensionality reduction (200 → 50 components)...')
pca = PCA(n_components=50, random_state=42)
X_pca = pca.fit_transform(X_scaled)
cumvar = np.cumsum(pca.explained_variance_ratio_)
print(f'  Variance explained by 50 PCs: {cumvar[-1]*100:.1f}%')
print(f'  Variance by first 10 PCs: {cumvar[9]*100:.1f}%')

# Scree plot
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(range(1, 26), pca.explained_variance_ratio_[:25] * 100,
        'o-', color='#0D9488', ms=5)
ax.set(xlabel='Principal Component', ylabel='Explained Variance (%)',
       title='PCA Scree Plot — Cell Painting Features')
ax.axvline(10, color='gray', lw=0.8, ls='--', alpha=0.5)
plt.tight_layout(); plt.show()
print('After PCA, the data is in a 50-dimensional space ready for clustering and UMAP.')
```

```python
# ============================================================
# CELL 4: UMAP visualisation of morphological profiles
# ============================================================
print('Step 3: UMAP projection (50 PCs → 2D)...')
reducer = umap.UMAP(n_neighbors=15, n_components=2, metric='euclidean',
                    random_state=42, min_dist=0.1)
X_umap = reducer.fit_transform(X_pca)
print(f'UMAP coordinates shape: {X_umap.shape}')

fig, ax = plt.subplots(figsize=(9, 7))
ax.set_facecolor('#F8FAFC')

for moa, color in MOA_COLORS.items():
    mask = df_profiles['true_moa'] == moa
    ax.scatter(X_umap[mask, 0], X_umap[mask, 1],
               c=color, s=40, alpha=0.75, label=moa.replace('_', ' '),
               edgecolors='white', lw=0.4)

ax.set(xlabel='UMAP-1', ylabel='UMAP-2',
       title='Cell Painting Profiles: UMAP Visualisation\n'
             'Compounds with similar mechanisms cluster together')
ax.legend(bbox_to_anchor=(1.01, 1), loc='upper left', fontsize=9)
ax.spines['top'].set_visible(False); ax.spines['right'].set_visible(False)
plt.tight_layout(); plt.show()

print('\nKey observation:')
print('Each mechanism of action forms a distinct cluster in morphological space.')
print('This means we can predict MoA for a new compound by finding its nearest cluster!')
```

```python
# ============================================================
# CELL 5: K-means clustering and MoA assignment
# ============================================================
print('Step 4: K-means clustering to identify mechanism clusters...')
print('We use k=5 (one cluster per expected mechanism) for this demo.')

k = 5  # number of MoA clusters
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
cluster_labels = kmeans.fit_predict(X_pca)

sil = silhouette_score(X_pca, cluster_labels)
print(f'  Silhouette score: {sil:.3f}  (0=random, 1=perfect separation)')

df_profiles['cluster'] = cluster_labels

# How well does clustering recover true MoA?
print('\nCluster → True MoA assignment:')
for c in range(k):
    subset = df_profiles[df_profiles['cluster'] == c]['true_moa']
    top_moa = subset.value_counts().index[0]
    purity  = subset.value_counts().iloc[0] / len(subset)
    print(f'  Cluster {c}: majority MoA = {top_moa:<25} (purity={purity:.2f})')

print('\n=== Notebook 6 Complete ===')

print('We will predict absorption, distribution, metabolism, excretion and toxicity.')
```
