# Notebook 5 — Virtual Screening for QSAR

---

## What you will learn
Virtual screening applies trained ML models to rank large compound libraries
by predicted potency — without running any lab experiments first.

### Workflow
1. Load a trained QSAR model (from Notebook 4)
2. Load a compound library (from ChEMBL or a file)
3. Apply Lipinski/ADMET pre-filters to remove obviously bad compounds
4. Compute fingerprints and descriptors for all compounds
5. Predict EGFR potency for every compound
6. Rank compounds by predicted pIC50 → top candidates for lab testing

## Why virtual screening?
- A typical drug discovery library has 10,000 to 10 million compounds
- Testing all experimentally costs millions of dollars and months of time
- Virtual screening narrows the list to the top 100–1000 candidates
- Hit rate in selected compounds is 5–20x higher than random selection

**Environment:** Google Colab or local Python (CPU sufficient)

```python
# ============================================================
# CELL 1: Install libraries
# ============================================================
print('Installing packages...')
import subprocess, sys
subprocess.run([sys.executable, '-m', 'pip', 'install', '--quiet',
                'rdkit', 'chembl-webresource-client',
                'xgboost', 'pandas', 'numpy', 'matplotlib'], check=False)
print('Done!')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

from rdkit import Chem, RDLogger
from rdkit.Chem import AllChem, Descriptors, rdMolDescriptors, DataStructs, Draw
RDLogger.DisableLog('rdApp.*')
import xgboost as xgb

print(f'NumPy   : {np.__version__}')
print(f'XGBoost : {xgb.__version__}')
print('Libraries imported!')
```

```python
# ============================================================
# CELL 2: Retrain a quick QSAR model (or load from Notebook 4)
# ============================================================
# In a real workflow you would load the saved model from Notebook 4.
# Here we retrain quickly so the notebook is self-contained.

from chembl_webresource_client.new_client import new_client
from rdkit.Chem.Scaffolds import MurckoScaffold
from collections import defaultdict
from sklearn.metrics import r2_score, mean_squared_error

print('Building QSAR model for EGFR (quick retraining)...')

def mol_to_features(smiles):
    """Convert SMILES to 1029-dim feature vector (1024 fp bits + 5 descriptors)."""
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        return None
    fp  = AllChem.GetMorganFingerprintAsBitVect(mol, 2, nBits=1024)
    arr = np.zeros(1024, dtype=np.float32)
    DataStructs.ConvertToNumpyArray(fp, arr)
    desc = np.array([
        Descriptors.MolWt(mol),
        Descriptors.MolLogP(mol),
        Descriptors.TPSA(mol),
        rdMolDescriptors.CalcNumHBD(mol),
        rdMolDescriptors.CalcNumHBA(mol),
    ], dtype=np.float32)
    return np.concatenate([arr, desc])

try:
    print('Fetching EGFR training data from ChEMBL...')
    activity = new_client.activity
    raw = activity.filter(target_chembl_id='CHEMBL203', standard_type='IC50',
                          standard_units='nM', standard_relation='='
                          ).only(['molecule_chembl_id','canonical_smiles','standard_value'])[:300]
    df_raw = pd.DataFrame(list(raw))
    df_raw['standard_value'] = pd.to_numeric(df_raw['standard_value'], errors='coerce')
    df_raw = df_raw.dropna(subset=['canonical_smiles','standard_value'])
    df_raw = df_raw[df_raw['standard_value'] > 0]
    df_dedup = (df_raw.groupby('molecule_chembl_id')
                .agg(smiles=('canonical_smiles','first'),
                     IC50_nM=('standard_value', lambda x: np.exp(np.log(x).mean())))
                .reset_index())
    df_dedup['pIC50'] = -np.log10(df_dedup['IC50_nM'] * 1e-9)
    rows = []
    for _, row in df_dedup.iterrows():
        feat = mol_to_features(row['smiles'])
        if feat is not None:
            rows.append({'smiles': row['smiles'], 'pIC50': row['pIC50'], 'feat': feat})
    df_model = pd.DataFrame(rows)
    X = np.stack(df_model['feat'])
    y = df_model['pIC50'].values.astype(np.float32)
    split = int(0.8 * len(X))
    model = xgb.XGBRegressor(n_estimators=200, max_depth=5, learning_rate=0.1,
                              random_state=42, verbosity=0, n_jobs=-1)
    model.fit(X[:split], y[:split])
    y_pred = model.predict(X[split:])
    r2  = r2_score(y[split:], y_pred)
    rmse = np.sqrt(mean_squared_error(y[split:], y_pred))
    print(f'Model trained: R²={r2:.3f}  RMSE={rmse:.3f} pIC50 units')
    TRAIN_SMILES = set(df_model['smiles'].values)
except Exception as e:
    print(f'ChEMBL failed ({e}). Using stub model.')
    model = xgb.XGBRegressor(n_estimators=50, random_state=42, verbosity=0)
    model.fit(np.random.rand(50, 1029), np.random.rand(50) * 3 + 4)
    TRAIN_SMILES = set()
print('QSAR model ready for virtual screening!')
```

```python
# ============================================================
# CELL 3: Load screening library
# ============================================================
# In real drug discovery, the library may contain millions of compounds
# (commercial vendors like Enamine, eMolecules, ZINC).
# Here we use ChEMBL approved drugs as a demo library.

print('=== Loading Screening Library ===')
print('Using ChEMBL approved drugs as a demo screening library')
print('In practice you would load a commercial vendor library (ZINC, Enamine, etc.)')

try:
    molecule = new_client.molecule
    # Get FDA-approved small molecules
    approved = molecule.filter(max_phase=4, molecule_type='Small molecule'
                               ).only(['molecule_chembl_id','pref_name',
                                       'molecule_structures'])[:200]
    lib_rows = []
    for m in approved:
        structs = m.get('molecule_structures')
        if structs and structs.get('canonical_smiles'):
            lib_rows.append({
                'chembl_id': m['molecule_chembl_id'],
                'name':      m.get('pref_name', 'Unknown'),
                'smiles':    structs['canonical_smiles'],
            })
    df_lib = pd.DataFrame(lib_rows)
    print(f'Loaded {len(df_lib)} compounds from ChEMBL approved drugs')
except Exception as e:
    print(f'ChEMBL failed ({e}). Using hard-coded demo library.')
    # Fallback: representative kinase inhibitors + non-kinase drugs
    demo_mols = [
        ('DEMO001', 'Erlotinib',   'COc1cc2ncnc(Nc3cccc(C#C)c3)c2cc1OCCO'),
        ('DEMO002', 'Gefitinib',   'COc1cc2ncnc(Nc3ccc(F)c(Cl)c3)c2cc1OCCCN1CCOCC1'),
        ('DEMO003', 'Imatinib',    'Cc1ccc(NC(=O)c2ccc(CN3CCN(C)CC3)cc2)cc1Nc1nccc(-c2cccnc2)n1'),
        ('DEMO004', 'Aspirin',     'CC(=O)Oc1ccccc1C(=O)O'),
        ('DEMO005', 'Metformin',   'CN(C)C(=N)NC(=N)N'),
        ('DEMO006', 'Atorvastatin','CC(C)c1c(C(=O)Nc2ccccc2)c(-c2ccccc2)c(-c2ccc(F)cc2)n1CCC(O)CC(O)CC(=O)O'),
        ('DEMO007', 'Tamoxifen',   'CCC(=C(c1ccccc1)c1ccc(OCCN(C)C)cc1)c1ccccc1'),
        ('DEMO008', 'Osimertinib', 'COc1cc2ncnc(Nc3cccc(NC(=O)/C=C/CN(C)C)c3)c2cc1OCCCN1CCOCC1'),
    ] * 10
    df_lib = pd.DataFrame(demo_mols, columns=['chembl_id', 'name', 'smiles'])
    df_lib = df_lib.drop_duplicates(subset='smiles').reset_index(drop=True)
    print(f'Using {len(df_lib)} demo compounds')

print(f'\nLibrary columns: {list(df_lib.columns)}')
print(df_lib.head(3).to_string(index=False))
```

```python
# ============================================================
# CELL 4: Apply Lipinski pre-filter and compute features
# ============================================================
# Pre-filtering removes compounds that are unlikely to be orally bioavailable.
# This reduces the screening library size before the expensive model inference step.

print('=== Step 1: Lipinski Pre-filter ===')
print('Removing compounds that violate > 1 Lipinski Rule of Five:')
print('  MW < 500 Da, HBD ≤ 5, HBA ≤ 10, LogP ≤ 5')

def passes_lipinski(smiles):
    """Returns True if compound passes Lipinski Ro5 (≤1 violation)."""
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        return False
    violations = sum([
        Descriptors.MolWt(mol) > 500,
        Descriptors.MolLogP(mol) > 5,
        rdMolDescriptors.CalcNumHBD(mol) > 5,
        rdMolDescriptors.CalcNumHBA(mol) > 10,
    ])
    return violations <= 1

n_before = len(df_lib)
df_lib['passes_ro5'] = df_lib['smiles'].apply(passes_lipinski)
df_filtered = df_lib[df_lib['passes_ro5']].copy().reset_index(drop=True)
n_after = len(df_filtered)
print(f'Before filter: {n_before} compounds')
print(f'After  filter: {n_after} compounds  ({n_after/n_before*100:.1f}% retained)')

# ---- Compute features for all filtered compounds ----
print('\n=== Step 2: Computing fingerprints and descriptors ===')
print(f'Processing {n_after} compounds...')

feats, valid_indices = [], []
for i, row in df_filtered.iterrows():
    feat = mol_to_features(row['smiles'])
    if feat is not None:
        feats.append(feat)
        valid_indices.append(i)

X_screen = np.stack(feats)
df_screen = df_filtered.loc[valid_indices].reset_index(drop=True)
print(f'Feature matrix: {X_screen.shape[0]} compounds × {X_screen.shape[1]} features')
```

```python
# ============================================================
# CELL 5: Score all compounds and rank
# ============================================================
print('=== Step 3: Predicting EGFR pIC50 for all compounds ===')

predicted_pIC50 = model.predict(X_screen)
df_screen['predicted_pIC50'] = predicted_pIC50
df_screen['predicted_IC50_nM'] = 10 ** (-predicted_pIC50) * 1e9

# Sort by predicted potency (highest pIC50 = most potent = top of list)
df_ranked = df_screen.sort_values('predicted_pIC50', ascending=False).reset_index(drop=True)

print(f'Screening complete! Ranked {len(df_ranked)} compounds.')
print('\n=== Top 15 Predicted EGFR Inhibitors ===')
top_cols = ['name', 'chembl_id', 'predicted_pIC50', 'predicted_IC50_nM']
print(df_ranked[top_cols].head(15).to_string(index=True))

# Distribution of predicted potency
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

axes[0].hist(predicted_pIC50, bins=25, color='#0D9488', edgecolor='white', alpha=0.85)
axes[0].axvline(6.0, color='red', lw=1.5, ls='--', label='pIC50=6 (1 µM)')
axes[0].set(xlabel='Predicted pIC50', ylabel='Count', title='Distribution of Predicted Potency')
axes[0].legend()

top20 = df_ranked.head(20)
axes[1].barh(range(20), top20['predicted_pIC50'].values[::-1],
             color='#0D9488', edgecolor='white')
axes[1].set_yticks(range(20))
axes[1].set_yticklabels(top20['name'].values[::-1], fontsize=7)
axes[1].set(xlabel='Predicted pIC50', title='Top 20 Predicted EGFR Inhibitors')
axes[1].axvline(6.0, color='red', lw=1.0, ls='--', alpha=0.7)

plt.tight_layout()
plt.show()

n_hits = (predicted_pIC50 >= 6).sum()
print(f'\nPredicted hits (pIC50 ≥ 6): {n_hits}/{len(df_ranked)} ({n_hits/len(df_ranked)*100:.1f}%)')
print('\nThese top compounds would be prioritised for experimental testing!')
print('\n=== Notebook 5 Complete ===')
```
