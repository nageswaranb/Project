## What you will learn
Scientific papers contain enormous amounts of drug discovery knowledge, but reading
and extracting that knowledge manually is slow and error-prone.
LLMs can help by:

1. **Named Entity Recognition (NER):** Extract drug names, targets, diseases from text
2. **Relation extraction:** Find drug-target, drug-indication, drug-AE relationships
3. **Summarisation:** Summarise papers to key findings in seconds
4. **Question answering:** Ask specific questions over paper PDFs
5. **RAG (Retrieval-Augmented Generation):** Search a corpus and answer with citations

## Use cases in drug discovery
- Competitive intelligence: What compounds are other groups working on for my target?
- Adverse event mining: What side effects are reported for a drug class?
- Mechanism of action: How does compound X work at the molecular level?
- Patent analysis: Is my compound structurally similar to existing patents?

````python
# ============================================================
# CELL 1: Install libraries
# ============================================================
print('Installing packages...')
import subprocess, sys
subprocess.run([sys.executable, '-m', 'pip', 'install', '--quiet',
                'anthropic', 'openai',
                'requests', 'beautifulsoup4',
                'pandas', 'numpy', 'matplotlib'], check=False)
print('Done!')

import os, re, json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import requests
import warnings
warnings.filterwarnings('ignore')

ANTHROPIC_API_KEY = os.environ.get('ANTHROPIC_API_KEY')
print(f'Anthropic key: {"found" if ANTHROPIC_API_KEY else "NOT found"}')
print('Libraries imported!')
````

````python
# ============================================================
# CELL 2: Fetch a PubMed abstract for analysis
# ============================================================
# PubMed provides free access to 35+ million biomedical abstracts via API.
# PMID (PubMed ID) is the unique identifier for each paper.

print('=== Step 1: Fetching PubMed Abstracts ===')
print('Using the PubMed E-utilities API (no API key required for < 10 req/sec)')

def fetch_pubmed_abstract(pmid):
    """
    Fetch a PubMed abstract by PMID (PubMed ID).
    Returns dict with title, abstract, authors, year.
    """
    url = f'https://pubmed.ncbi.nlm.nih.gov/{pmid}/?format=abstract'
    try:
        from bs4 import BeautifulSoup
        response = requests.get(url, timeout=10)
        soup = BeautifulSoup(response.text, 'html.parser')

        title  = soup.find('h1', class_='heading-title')
        title  = title.get_text(strip=True) if title else 'Title not found'

        abstract = soup.find('div', class_='abstract-content')
        abstract = abstract.get_text(strip=True) if abstract else 'Abstract not found'

        return {'pmid': pmid, 'title': title, 'abstract': abstract}
    except Exception as e:
        return {'pmid': pmid, 'title': f'Fetch failed: {e}',
                'abstract': ''}

# Example PubMed IDs for EGFR inhibitor papers
PMIDS = [
    '24887559',   # Osimertinib (AZD9291) discovery paper
    '31810215',   # EGFR resistance mechanisms review
]

papers = []
for pmid in PMIDS:
    print(f'  Fetching PMID {pmid}...')
    paper = fetch_pubmed_abstract(pmid)
    papers.append(paper)
    print(f'  Title: {paper["title"][:70]}...')
    print(f'  Abstract preview: {paper["abstract"][:100]}...')
    print()

print(f'Fetched {len(papers)} abstracts')
````

````python
# ============================================================
# CELL 3: Named Entity Recognition with an LLM
# ============================================================
# NER = identifying and classifying entities in text.
# Drug discovery entities: compound names, targets, diseases, genes, assays.

print('=== Step 2: Named Entity Recognition (NER) ===')
print('Extracting drug names, targets, diseases, and genes from abstract text.\n')

# Demo abstract (Osimertinib-related, in case PubMed fetch failed)
DEMO_ABSTRACT = """AZD9291 is an irreversible, mutant-selective epidermal growth factor receptor (EGFR)
inhibitor designed to overcome resistance to first-generation EGFR inhibitors such as erlotinib and
gefitinib in non-small cell lung cancer (NSCLC) patients. The T790M gatekeeper mutation in EGFR
accounts for approximately 50-60% of acquired resistance cases. AZD9291 (osimertinib) maintains
selectivity over wild-type EGFR by forming a covalent bond with Cys797 in the ATP-binding pocket.
In phase I/II clinical trials (AURA study), AZD9291 demonstrated a 61% objective response rate
in EGFR T790M-positive patients. Common adverse effects included rash, diarrhoea, and nausea."""

TEXT_TO_ANALYZE = papers[0]['abstract'] if papers[0]['abstract'] != 'Abstract not found' else DEMO_ABSTRACT
print(f'Text to analyse ({len(TEXT_TO_ANALYZE)} characters):')
print(TEXT_TO_ANALYZE[:300] + '...' if len(TEXT_TO_ANALYZE) > 300 else TEXT_TO_ANALYZE)

if ANTHROPIC_API_KEY:
    import anthropic
    client = anthropic.Anthropic(api_key=ANTHROPIC_API_KEY)

    ner_prompt = f"""Extract all named entities from this biomedical text and classify them.
Return a JSON object with these categories:
  drugs: list of drug/compound names
  targets: list of protein/gene targets
  diseases: list of diseases/conditions
  mutations: list of genetic mutations
  assays: list of assay/study types

TEXT:
{TEXT_TO_ANALYZE}

Return ONLY valid JSON, no other text."""

    print('\nSending to Claude Sonnet for NER...')
    response = client.messages.create(
        model='claude-sonnet-4-5', max_tokens=500,
        messages=[{'role': 'user', 'content': ner_prompt}]
    )
    ner_text = response.content[0].text.strip()

    try:
        # Clean up response (sometimes LLMs add markdown code blocks)
        ner_text_clean = re.sub(r'^```json\s*', '', ner_text, flags=re.MULTILINE)
        ner_text_clean = re.sub(r'```$', '', ner_text_clean, flags=re.MULTILINE).strip()
        entities = json.loads(ner_text_clean)
        print('\n=== Extracted Named Entities ===')
        for category, items in entities.items():
            print(f'  {category}: {items}')
    except json.JSONDecodeError:
        print('Raw response from Claude:')
        print(ner_text)

else:
    print('\nNo API key — showing expected output format:')
    example_entities = {
        'drugs':     ['AZD9291', 'osimertinib', 'erlotinib', 'gefitinib'],
        'targets':   ['EGFR', 'epidermal growth factor receptor'],
        'diseases':  ['non-small cell lung cancer', 'NSCLC'],
        'mutations': ['T790M', 'Cys797'],
        'assays':    ['phase I/II clinical trials', 'AURA study'],
    }
    for cat, items in example_entities.items():
        print(f'  {cat}: {items}')
````

````python
# ============================================================
# CELL 4: Structured data extraction — drug-target relationships
# ============================================================

print('=== Step 3: Extract Drug-Target Relationships ===')
print('Converting unstructured text into structured database entries.\n')

MULTI_ABSTRACT = """
PAPER 1: Erlotinib (OSI-774) demonstrated IC50 of 2 nM against EGFR and showed
clinical activity in NSCLC patients harbouring EGFR activating mutations (L858R, del19).

PAPER 2: The EGFR T790M gatekeeper mutation confers resistance to erlotinib and gefitinib.
Third-generation inhibitors such as osimertinib (AZD9291) overcome this resistance
by covalently binding to Cys797 and showing 200-fold selectivity over wild-type EGFR.

PAPER 3: Alectinib targets ALK (anaplastic lymphoma kinase) with IC50 of 1.9 nM and
is approved for ALK-rearranged NSCLC. It also inhibits RET with IC50 of 4.8 nM.
"""

if ANTHROPIC_API_KEY:
    extract_prompt = f"""Extract all drug-target interaction data from these biomedical texts.
Return a JSON array where each entry has:
  drug: compound name
  target: protein/gene target
  ic50_nM: IC50 value in nM (null if not stated)
  indication: disease indication
  notes: any additional details (mutations, selectivity, etc.)

TEXTS:
{MULTI_ABSTRACT}

Return ONLY valid JSON array."""

    response = client.messages.create(
        model='claude-sonnet-4-5', max_tokens=600,
        messages=[{'role': 'user', 'content': extract_prompt}]
    )
    extract_text = response.content[0].text.strip()
    extract_clean = re.sub(r'^```json\s*', '', extract_text, flags=re.MULTILINE)
    extract_clean = re.sub(r'```$', '', extract_clean, flags=re.MULTILINE).strip()

    try:
        relations = json.loads(extract_clean)
        df_relations = pd.DataFrame(relations)
        print('Extracted Drug-Target Database:')
        print(df_relations.to_string(index=False))
        df_relations.to_csv('drug_target_relations.csv', index=False)
        print('\nSaved to drug_target_relations.csv')
    except Exception as e:
        print(f'Parse error: {e}\nRaw response:\n{extract_text}')
else:
    print('Expected output:')
    example = pd.DataFrame([
        {'drug': 'Erlotinib', 'target': 'EGFR', 'ic50_nM': 2.0, 'indication': 'NSCLC', 'notes': 'L858R, del19 mutations'},
        {'drug': 'Osimertinib', 'target': 'EGFR', 'ic50_nM': None, 'indication': 'NSCLC', 'notes': 'T790M resistant, covalent C797'},
        {'drug': 'Alectinib', 'target': 'ALK', 'ic50_nM': 1.9, 'indication': 'NSCLC', 'notes': 'ALK rearrangement'},
        {'drug': 'Alectinib', 'target': 'RET', 'ic50_nM': 4.8, 'indication': 'NSCLC', 'notes': 'Off-target activity'},
    ])
    print(example.to_string(index=False))

print('\n=== Notebook 9 Complete ===')
````
