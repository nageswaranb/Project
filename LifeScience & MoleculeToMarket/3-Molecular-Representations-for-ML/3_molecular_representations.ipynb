## What you will learn
Machine learning models work with numbers. Molecules are chemical structures.
This notebook covers the three main ways to convert molecules to ML-ready numbers:

1. **SMILES** — the universal text language for molecular structure
2. **Fingerprints** — binary bit vectors encoding structural features
   - Morgan (circular) fingerprints — the most popular for ML
   - MACCS keys — 166-bit expert-defined structural keys
3. **Physicochemical descriptors** — calculated properties (MW, logP, TPSA, etc.)
4. **Lipinski Rule of Five** — drug-likeness filter based on physicochemical properties
5. **ChEMBL integration** — building a real EGFR inhibitor dataset for QSAR

## Why representations matter
The choice of representation directly impacts model performance:
- Morgan fingerprints capture local chemical environments
- Descriptors capture global physicochemical properties
- Each encodes different aspects of the molecule's chemical space
- For QSAR models, using BOTH together often gives the best results

## Cell 2: SMILES — the language of molecules

**SMILES** (Simplified Molecular Input Line Entry System) encodes a 3D molecule as a 1D string.
Rules:
- Atoms: `C` carbon, `N` nitrogen, `O` oxygen, `c` aromatic carbon
- Bonds: `-` single, `=` double, `#` triple, `:` aromatic (implicit)
- Branches: `()` parentheses
- Rings: numbers indicate ring closures, e.g. `C1CCCCC1` = cyclohexane

Example drugs we will work with:
- **Erlotinib** (Tarceva) — 1st-gen EGFR inhibitor
- **Gefitinib** (Iressa) — 1st-gen EGFR inhibitor
- **Osimertinib** (Tagrisso) — 3rd-gen EGFR inhibitor (overcomes T790M)
- **Aspirin** — common NSAID (non-kinase drug, used as comparison)

```python
# ============================================================
# CELL 2: SMILES — parsing molecules from strings
# ============================================================

# SMILES strings for well-known EGFR inhibitors + control drugs
DRUG_SMILES = {
    'Erlotinib':   'COCCOC1=CC=C2C(=C1OCC)NC(=NC3=CC=CC=C3)N=C2',   # 1st-gen EGFR inhibitor
    'Gefitinib':   'COC1=CC2=C(C=C1OCCCN3CCOCC3)C(=NC=N2)NC4=CC=C(Cl)C=C4F',  # 1st-gen EGFR inhibitor
    'Osimertinib': 'COc1cc2ncnc(Nc3cccc(NC(=O)/C=C/CN(C)C)c3)c2cc1OCCCN1CCOCC1',  # 3rd-gen
    'Aspirin':     'CC(=O)Oc1ccccc1C(=O)O',                           # non-EGFR reference
    'Caffeine':    'CN1C=NC2=C1C(=O)N(C)C(=O)N2C',                   # non-drug reference
}

print('=== SMILES Parsing with RDKit ===')
print('RDKit converts SMILES strings into internal molecular graph objects.\n')

mols = {}
for name, smiles in DRUG_SMILES.items():
    mol = Chem.MolFromSmiles(smiles)   # parse SMILES → RDKit Mol object
    if mol is None:
        print(f'  FAILED to parse: {name} ({smiles})')
    else:
        mols[name] = mol
        # Count atoms and bonds
        n_atoms = mol.GetNumAtoms()
        n_bonds = mol.GetNumBonds()
        n_rings  = rdMolDescriptors.CalcNumRings(mol)
        print(f'  {name:<15}: {n_atoms:3d} atoms  {n_bonds:3d} bonds  {n_rings} rings  |  SMILES: {smiles[:40]}...')

print(f'\nSuccessfully parsed {len(mols)}/{len(DRUG_SMILES)} molecules')
print('\nCanonical SMILES (standardised form RDKit uses internally):')
for name, mol in mols.items():
    canonical = Chem.MolToSmiles(mol)  # convert back to canonical SMILES
    print(f'  {name}: {canonical[:60]}...')
```

```python
# ============================================================
# CELL 3: Morgan (circular) fingerprints
# ============================================================
# Morgan fingerprints encode the local chemical environment around each atom
# using a circular algorithm out to radius r bonds.
# radius=2 is standard for drug discovery (captures 2-bond neighbourhoods).
# nBits=2048 is a common vector length.

print('=== Morgan Fingerprints ===')
print('Algorithm: For each atom, hash the circular neighbourhood of radius r')
print('Standard: radius=2, nBits=2048 (each bit = one structural feature)\n')

RADIUS   = 2
N_BITS   = 2048

morgan_fps = {}
for name, mol in mols.items():
    # Generate bit vector fingerprint
    fp = AllChem.GetMorganFingerprintAsBitVect(mol, radius=RADIUS, nBits=N_BITS)
    morgan_fps[name] = fp
    # Count how many bits are set (1 vs 0)
    n_on = fp.GetNumOnBits()
    arr  = np.zeros(N_BITS, dtype=np.int8)
    DataStructs.ConvertToNumpyArray(fp, arr)
    print(f'  {name:<15}: {n_on:3d}/{N_BITS} bits ON ({n_on/N_BITS*100:.1f}% density)')

# Convert fingerprints to numpy matrix for ML
fp_matrix = np.zeros((len(mols), N_BITS), dtype=np.int8)
for i, (name, fp) in enumerate(morgan_fps.items()):
    DataStructs.ConvertToNumpyArray(fp, fp_matrix[i])

print(f'\nFingerprint matrix shape: {fp_matrix.shape}')
print('Each row = one molecule; each column = one structural feature bit')
print('This matrix can be directly fed into any sklearn ML model!')

# ---- Pairwise Tanimoto (Jaccard) similarity ----
print('\n=== Tanimoto Similarity between EGFR drugs ===')
print('Tanimoto = bits in common / (bits in A + bits in B - bits in common)')
print('1.0 = identical, 0.0 = no shared features\n')
drug_names = list(mols.keys())
for i in range(len(drug_names)):
    for j in range(i+1, len(drug_names)):
        sim = DataStructs.TanimotoSimilarity(morgan_fps[drug_names[i]],
                                             morgan_fps[drug_names[j]])
        print(f'  {drug_names[i]:<15} vs {drug_names[j]:<15}: Tanimoto = {sim:.3f}')
```

```python
# ============================================================
# CELL 4: Physicochemical descriptors — the Lipinski Rule of Five
# ============================================================
# Lipinski's Rule of Five (Ro5) predicts oral bioavailability:
#   MW < 500 Da
#   HBD (H-bond donors) ≤ 5
#   HBA (H-bond acceptors) ≤ 10
#   logP (lipophilicity) ≤ 5
# Compounds violating > 1 rule are likely to have poor oral absorption.

print('=== Physicochemical Descriptors ===')
print('Calculating Lipinski Rule of Five (Ro5) properties for each molecule\n')

descriptor_data = []
for name, mol in mols.items():
    mw   = Descriptors.MolWt(mol)                              # molecular weight
    logp = Descriptors.MolLogP(mol)                            # Wildman-Crippen logP
    hbd  = rdMolDescriptors.CalcNumHBD(mol)                   # H-bond donors
    hba  = rdMolDescriptors.CalcNumHBA(mol)                   # H-bond acceptors
    tpsa = Descriptors.TPSA(mol)                              # topological polar surface area
    rotbonds = rdMolDescriptors.CalcNumRotatableBonds(mol)    # rotatable bonds (flexibility)
    rings    = rdMolDescriptors.CalcNumRings(mol)
    arom     = rdMolDescriptors.CalcNumAromaticRings(mol)

    # Lipinski rule check: count violations
    violations = sum([
        mw > 500,
        hbd > 5,
        hba > 10,
        logp > 5,
    ])
    drug_like = 'YES' if violations <= 1 else 'NO'

    descriptor_data.append({
        'Compound':     name,
        'MW (Da)':      round(mw, 1),
        'LogP':         round(logp, 2),
        'HBD':          hbd,
        'HBA':          hba,
        'TPSA (Å²)':    round(tpsa, 1),
        'Rot. Bonds':   rotbonds,
        'Rings':        rings,
        'Arom. Rings':  arom,
        'Ro5 Violations': violations,
        'Drug-like':    drug_like,
    })
    print(f'  {name}:')
    print(f'    MW={mw:.1f} Da  logP={logp:.2f}  HBD={hbd}  HBA={hba}  TPSA={tpsa:.1f} Å²')
    print(f'    Ro5 violations: {violations}  → Drug-like: {drug_like}')

desc_df = pd.DataFrame(descriptor_data)
print(f'\nDescriptor table ({desc_df.shape[0]} compounds × {desc_df.shape[1]} properties):')
print(desc_df.to_string(index=False))
print('\nNote: All EGFR inhibitors should pass Ro5 (they are oral drugs)')
```

```python
# ============================================================
# CELL 5: Pull EGFR dataset from ChEMBL and build ML-ready table
# ============================================================
# ChEMBL contains experimentally measured IC50 values for drug-like compounds.
# We will:
#   1. Pull EGFR IC50 data
#   2. Compute Morgan fingerprints for each compound
#   3. Compute physicochemical descriptors
#   4. Combine into a final ML-ready DataFrame

from chembl_webresource_client.new_client import new_client

print('Querying ChEMBL for EGFR (CHEMBL203) IC50 data...')

try:
    activity = new_client.activity
    egfr_raw = activity.filter(
        target_chembl_id='CHEMBL203',
        standard_type='IC50',
        standard_units='nM',
        standard_relation='=',
    ).only(['molecule_chembl_id', 'canonical_smiles', 'standard_value', 'standard_units'])[:200]

    egfr_df = pd.DataFrame(list(egfr_raw))
    egfr_df['standard_value'] = pd.to_numeric(egfr_df['standard_value'], errors='coerce')
    egfr_df = egfr_df.dropna(subset=['canonical_smiles', 'standard_value'])
    egfr_df = egfr_df.drop_duplicates(subset='molecule_chembl_id')
    egfr_df = egfr_df[egfr_df['standard_value'] > 0].reset_index(drop=True)

    print(f'Retrieved {len(egfr_df)} unique EGFR IC50 records')
    print(f'IC50 range: {egfr_df["standard_value"].min():.1f} to {egfr_df["standard_value"].max():.0f} nM')

    # Compute pIC50 (−log10(IC50 in M)): higher pIC50 = more potent
    egfr_df['pIC50'] = -np.log10(egfr_df['standard_value'] * 1e-9)

    # Parse SMILES and compute fingerprints
    print('\nComputing Morgan fingerprints and descriptors...')
    valid_rows = []
    for _, row in egfr_df.iterrows():
        mol = Chem.MolFromSmiles(row['canonical_smiles'])
        if mol is None:
            continue
        fp  = AllChem.GetMorganFingerprintAsBitVect(mol, radius=2, nBits=1024)
        arr = np.zeros(1024, dtype=np.int8)
        DataStructs.ConvertToNumpyArray(fp, arr)
        mw   = Descriptors.MolWt(mol)
        logp = Descriptors.MolLogP(mol)
        tpsa = Descriptors.TPSA(mol)
        hbd  = rdMolDescriptors.CalcNumHBD(mol)
        hba  = rdMolDescriptors.CalcNumHBA(mol)
        valid_rows.append({
            'chembl_id':   row['molecule_chembl_id'],
            'smiles':      row['canonical_smiles'],
            'IC50_nM':     row['standard_value'],
            'pIC50':       row['pIC50'],
            'MW':          mw,
            'LogP':        logp,
            'TPSA':        tpsa,
            'HBD':         hbd,
            'HBA':         hba,
            **{f'fp_{i}': arr[i] for i in range(1024)},
        })

    final_df = pd.DataFrame(valid_rows)
    print(f'\nFinal ML-ready dataset: {final_df.shape[0]} compounds × {final_df.shape[1]} columns')
    print('  Columns: chembl_id, smiles, IC50_nM, pIC50, MW, LogP, TPSA, HBD, HBA, fp_0..fp_1023')

    # Save for notebook 4
    final_df.to_csv('egfr_ml_dataset.csv', index=False)
    print('\nSaved egfr_ml_dataset.csv for use in Notebook 4 (QSAR Modeling)')

    # pIC50 distribution plot
    fig, ax = plt.subplots(figsize=(7, 4))
    ax.hist(final_df['pIC50'], bins=25, color='#0D9488', edgecolor='white', alpha=0.85)
    ax.axvline(6.0, color='red', lw=1.5, ls='--', label='pIC50=6 (1 µM threshold)')
    ax.set(xlabel='pIC50 (higher = more potent)', ylabel='Count',
           title='Distribution of EGFR Inhibitor Potency from ChEMBL')
    ax.legend()
    plt.tight_layout()
    plt.show()
    n_active   = (final_df['pIC50'] >= 6).sum()
    n_inactive = (final_df['pIC50'] < 6).sum()
    print(f'\nActive (pIC50 ≥ 6, IC50 ≤ 1µM): {n_active} compounds ({n_active/len(final_df)*100:.1f}%)')
    print(f'Inactive (pIC50 < 6)           : {n_inactive} compounds ({n_inactive/len(final_df)*100:.1f}%)')

except Exception as e:
    print(f'ChEMBL query failed: {e}')
    print('This may be a network issue. The dataset will be fetched again in Notebook 4.')

print('\n=== Notebook 3 Complete ===')
```
