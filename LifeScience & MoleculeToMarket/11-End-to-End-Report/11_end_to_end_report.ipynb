# Notebook 11 — End-to-End Drug Discovery Report

## What you will learn
This notebook integrates ALL the previous computational analyses into a single
end-to-end pipeline and generates an automated drug discovery report.

### Integrated pipeline
1. Target identification and characterisation (EGFR, from protein language model)
2. Compound screening and IC50 fitting (from Notebooks 2–4)
3. ADMET property assessment (from Notebook 7)
4. Molecular docking scores (from Notebook 10)
5. Literature evidence (from Notebook 9)
6. **Final compound prioritisation and PDF/HTML report generation**

## Why an integrated report?
Drug discovery decisions require combining evidence from multiple sources:
- Potency (IC50) alone is not enough
- Need ADMET, selectivity, docking, and literature all together
- An automated report makes it easy to compare and communicate across teams

```python
# ============================================================
# CELL 1: Install libraries
# ============================================================
print('Installing packages...')
import subprocess, sys
subprocess.run([sys.executable, '-m', 'pip', 'install', '--quiet',
                'rdkit', 'pandas', 'numpy', 'matplotlib', 'anthropic'], check=False)
print('Done!')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import os, warnings
warnings.filterwarnings('ignore')

from rdkit import Chem, RDLogger
from rdkit.Chem import AllChem, Descriptors, rdMolDescriptors, Draw, DataStructs
RDLogger.DisableLog('rdApp.*')

print('Libraries imported!')
```

```python
# ============================================================
# CELL 2: Assemble compound data from all previous analyses
# ============================================================
print('=== Assembling compound data from all analysis stages ===')
print()

# In a real pipeline you would load results saved from each previous notebook.
# Here we define representative data for 6 candidate EGFR inhibitors.

COMPOUNDS = {
    'Osimertinib': {
        'smiles':     'COc1cc2ncnc(Nc3cccc(NC(=O)/C=C/CN(C)C)c3)c2cc1OCCCN1CCOCC1',
        'IC50_nM':    1.0,
        'pIC50':      9.0,
        'docking':    -12.1,
        'BBB':        False,
        'solubility': 'moderate',
        'hERG_risk':  'low',
        'covalent':   True,
        'selectivity':'T790M selective',
        'status':     'FDA approved (2015)',
    },
    'Erlotinib': {
        'smiles':     'COc1cc2ncnc(Nc3cccc(C#C)c3)c2cc1OCCO',
        'IC50_nM':    2.0,
        'pIC50':      8.7,
        'docking':    -11.2,
        'BBB':        False,
        'solubility': 'low',
        'hERG_risk':  'low',
        'covalent':   False,
        'selectivity':'WT + L858R, fails T790M',
        'status':     'FDA approved (2004)',
    },
    'Gefitinib': {
        'smiles':     'COc1cc2ncnc(Nc3ccc(F)cc3Cl)c2cc1OCCCN1CCOCC1',
        'IC50_nM':    33.0,
        'pIC50':      7.5,
        'docking':    -10.8,
        'BBB':        False,
        'solubility': 'low',
        'hERG_risk':  'low',
        'covalent':   False,
        'selectivity':'WT + L858R, fails T790M',
        'status':     'FDA approved (2003)',
    },
    'Compound_X': {
        'smiles':     'CC(=O)Nc1ccc2c(ncnc2c1O)Nc1ccc(F)cc1',
        'IC50_nM':    5.0,
        'pIC50':      8.3,
        'docking':    -11.5,
        'BBB':        False,
        'solubility': 'high',
        'hERG_risk':  'moderate',
        'covalent':   False,
        'selectivity':'WT, L858R, T790M (hypothetical)',
        'status':     'virtual compound',
    },
    'Compound_Y': {
        'smiles':     'C=CC(=O)Nc1cccc(Nc2ncnc3cc(OCCN4CCOCC4)c(OC)cc23)c1',
        'IC50_nM':    12.0,
        'pIC50':      7.9,
        'docking':    -10.2,
        'BBB':        False,
        'solubility': 'moderate',
        'hERG_risk':  'low',
        'covalent':   True,
        'selectivity':'T790M selective (hypothetical)',
        'status':     'virtual compound',
    },
    'Aspirin_ctrl': {
        'smiles':     'CC(=O)Oc1ccccc1C(=O)O',
        'IC50_nM':    10000,
        'pIC50':      5.0,
        'docking':    -6.3,
        'BBB':        True,
        'solubility': 'high',
        'hERG_risk':  'low',
        'covalent':   False,
        'selectivity':'None (negative control)',
        'status':     'negative control',
    },
}

print(f'Compounds in report: {list(COMPOUNDS.keys())}')
print(f'\nData fields per compound:')
for key in list(COMPOUNDS.values())[0].keys():
    print(f'  {key}')
```

```python
# ============================================================
# CELL 3: Compute summary scores and prioritise compounds
# ============================================================
# We build a composite score combining potency + docking + ADMET.
# This converts multi-dimensional data into a single ranking metric.

print('=== Computing Composite Drug Discovery Score ===')
print()
print('Scoring weights:')
print('  Potency  (pIC50)  : 40%  — most important for a kinase inhibitor')
print('  Docking score     : 25%  — structural evidence for binding')
print('  ADMET (hERG risk) : 20%  — safety filter (hERG → cardiac arrhythmia)')
print('  Solubility        : 15%  — formulation and oral absorption')
print()

HERGRISK_SCORE = {'low': 1.0, 'moderate': 0.5, 'high': 0.0}
SOL_SCORE      = {'high': 1.0, 'moderate': 0.6, 'low': 0.3}

rows = []
for name, props in COMPOUNDS.items():
    mol = Chem.MolFromSmiles(props['smiles'])
    mw   = Descriptors.MolWt(mol) if mol else 0
    logp = Descriptors.MolLogP(mol) if mol else 0

    # Normalise pIC50 (5 to 10 range → 0 to 1)
    pot_score  = max(0, min(1, (props['pIC50'] - 5) / 5))
    # Normalise docking (−6 to −13 range → 0 to 1)
    dock_score = max(0, min(1, (-props['docking'] - 6) / 7))
    herg_score = HERGRISK_SCORE.get(props['hERG_risk'], 0.5)
    sol_score  = SOL_SCORE.get(props['solubility'], 0.5)

    composite  = (0.40 * pot_score + 0.25 * dock_score +
                  0.20 * herg_score + 0.15 * sol_score)

    rows.append({
        'Compound':          name,
        'IC50 (nM)':         props['IC50_nM'],
        'pIC50':             round(props['pIC50'], 2),
        'Docking (kcal/mol)':props['docking'],
        'MW (Da)':           round(mw, 0),
        'LogP':              round(logp, 2),
        'hERG risk':         props['hERG_risk'],
        'Solubility':        props['solubility'],
        'Covalent':          props['covalent'],
        'Status':            props['status'],
        'Composite Score':   round(composite, 3),
    })

df_report = pd.DataFrame(rows).sort_values('Composite Score', ascending=False)

print('=== Compound Prioritisation Table (sorted by composite score) ===')
print(df_report.to_string(index=False))
df_report.to_csv('compound_report.csv', index=False)
print('\nSaved compound_report.csv')
```

```python
# ============================================================
# CELL 4: Generate summary visualisation
# ============================================================
print('Generating end-to-end drug discovery summary figure...')

fig = plt.figure(figsize=(15, 10))
fig.suptitle('End-to-End EGFR Drug Discovery Report\n'
             'Multi-parameter compound prioritisation', 
             fontsize=14, fontweight='bold')

gs = gridspec.GridSpec(2, 3, hspace=0.4, wspace=0.4)

compounds = df_report['Compound'].values
colors    = ['#0D9488' if s in ['FDA approved (2015)', 'FDA approved (2004)', 'FDA approved (2003)']
             else '#1B2A4A' if 'hypothetical' in s
             else '#6B7280' for s in df_report['Status']]

# Panel 1: pIC50 bar chart
ax1 = fig.add_subplot(gs[0, 0])
ax1.barh(compounds[::-1], df_report['pIC50'].values[::-1], color=colors[::-1], edgecolor='white')
ax1.axvline(6, color='red', lw=1, ls='--', alpha=0.5)
ax1.set(xlabel='pIC50', title='Potency (pIC50)')

# Panel 2: Docking scores
ax2 = fig.add_subplot(gs[0, 1])
ax2.barh(compounds[::-1], df_report['Docking (kcal/mol)'].values[::-1], color=colors[::-1], edgecolor='white')
ax2.set(xlabel='Vina Score (kcal/mol)', title='Docking Score')

# Panel 3: Composite score
ax3 = fig.add_subplot(gs[0, 2])
ax3.barh(compounds[::-1], df_report['Composite Score'].values[::-1],
         color=['gold' if i == 0 else '#1B2A4A' for i in range(len(compounds))[::-1]],
         edgecolor='white')
ax3.set(xlabel='Composite Score (0-1)', title='Overall Priority Score')

# Panel 4: MW vs LogP scatter
ax4 = fig.add_subplot(gs[1, 0])
ax4.scatter(df_report['MW (Da)'], df_report['LogP'], s=100, c='#0D9488', alpha=0.8)
for _, row in df_report.iterrows():
    ax4.annotate(row['Compound'][:6], (row['MW (Da)'], row['LogP']),
                 textcoords='offset points', xytext=(4, 4), fontsize=7)
ax4.axvline(500, color='red', lw=0.8, ls='--', alpha=0.5, label='MW=500')
ax4.axhline(5, color='red', lw=0.8, ls='--', alpha=0.5, label='LogP=5')
ax4.set(xlabel='MW (Da)', ylabel='LogP', title='Chemical Space (Lipinski)')
ax4.legend(fontsize=7)

# Panel 5: IC50 bar
ax5 = fig.add_subplot(gs[1, 1])
ic50_vals = df_report['IC50 (nM)'].values
ax5.barh(compounds[::-1], np.log10(ic50_vals[::-1]), color=colors[::-1], edgecolor='white')
ax5.axvline(np.log10(1000), color='red', lw=1, ls='--', alpha=0.5)
ax5.set(xlabel='log10(IC50 nM)', title='Potency (log IC50)')

# Panel 6: Summary table
ax6 = fig.add_subplot(gs[1, 2])
ax6.axis('off')
top3 = df_report.head(3)
table_data = [['#', 'Compound', 'Score', 'Status']]
for i, row in enumerate(top3.itertuples()):
    table_data.append([str(i+1), row.Compound[:12], f'{row._13:.3f}', row.Status[:18]])
table = ax6.table(cellText=table_data[1:], colLabels=table_data[0],
                  loc='center', cellLoc='center')
table.auto_set_font_size(False)
table.set_fontsize(8)
table.scale(1, 1.5)
ax6.set_title('Top 3 Compounds', fontweight='bold')

plt.savefig('drug_discovery_report.png', dpi=150, bbox_inches='tight')
plt.show()
print('Report figure saved: drug_discovery_report.png')
print('\n=== Notebook 11 Complete ===')
```
