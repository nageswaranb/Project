## Protein Language Models: Mapping EGFR Mutations in Embedding Space

---

### What this notebook demonstrates

This notebook shows how **ESM2**, a protein language model from Meta AI (published in *Science* 2023 and trained on 250 million protein sequences), can represent protein sequences as mathematical vectors called **embeddings**.

We use EGFR (Epidermal Growth Factor Receptor) — a major cancer drug target — and its clinically important drug-resistance mutations:

- **T790M**: gatekeeper mutation — a threonine → methionine change at position 790. The bulky methionine fills the ATP binding pocket and prevents 1st-generation inhibitors (erlotinib, gefitinib) from binding.
- **C797S**: covalent anchor mutation — a cysteine → serine change at position 797. Osimertinib forms a covalent bond with C797; when C797 becomes S797, that bond cannot form.
- **L858R**: activating mutation — a leucine → arginine change at position 858. Makes EGFR constitutively active (drives tumor growth) but also sensitises the cancer to EGFR inhibitors.

### Key questions
- How does each single amino acid change shift EGFR in ESM2's learned protein space?
- Can we see the biological gradient: EGFR variants → HER2 (same family) → KRAS (different family) → random sequence?
- Can ESM2 score mutations zero-shot (without any disease training data)?

### Pipeline
1. Install libraries and load ESM2 from local path or HuggingFace
2. Fetch the EGFR kinase domain sequence from UniProt (P00533) and build mutants
3. Define the full protein panel (EGFR variants + HER2 + KRAS + random control)
4. Embed all proteins with ESM2
5. Pairwise cosine similarity heatmap
6. UMAP visualisation of protein embedding space
7. Mutation distance bar chart
8. Zero-shot variant effect prediction using the ESM2 MLM head

**Model:** `facebook/esm2_t33_650M_UR50D` (~2.6 GB — downloads automatically from HuggingFace if not cached locally)

## Cell 1: Install libraries and load ESM2

**What is ESM2?**

ESM2 is a transformer model trained as a masked language model on protein sequences.
Just as BERT for text learns grammar and meaning by predicting masked words,
ESM2 learns protein structure, function, and evolution by predicting masked amino acids.

The model produces a 1280-dimensional embedding vector for each residue.
Mean-pooling these gives one vector per protein — its position in ESM2's learned protein space.

**Key imports:**
- `EsmTokenizer`: converts amino acid letters to integer token IDs
- `EsmModel`: the transformer that produces embeddings (no prediction head)
- `EsmForMaskedLM`: same transformer + a prediction head (used in Cell 8 for mutation scoring)

**Device selection:**
- GPU: runs in float16 to save VRAM (~3 GB)
- CPU: runs in float32, is slower but still works

## Cell 2: Fetch EGFR kinase domain and build mutants

**Why the kinase domain?**

Full EGFR is 1210 residues. The kinase domain (residues 712–979) is where all the
drug-binding happens. Using just this 268-residue segment keeps embeddings efficient
and focuses attention on the clinically relevant part.

**Mutation indexing:**
UniProt uses 1-based residue numbers. In Python strings (0-indexed):
- Residue 790 (1-indexed) = kinase_domain[790 - 712] = kinase_domain[78]
- Residue 797 → kinase_domain[85]
- Residue 858 → kinase_domain[146]

We verify the expected wild-type amino acid before mutating each position.

```python
# ============================================================
# CELL 2: Fetch EGFR kinase domain and build mutants
# ============================================================

from Bio import Entrez, SeqIO

# NCBI Entrez requires an email (not validated, just for contact tracking)
Entrez.email = 'your@email.com'

EGFR_ACCESSION = 'P00533'   # human EGFR (ErbB-1) on UniProt
KINASE_START   = 712        # 1-indexed, inclusive
KINASE_END     = 979        # 1-indexed, inclusive

print(f'Fetching EGFR protein sequence from UniProt...')
print(f'  Accession    : {EGFR_ACCESSION}')
print(f'  Kinase domain: residues {KINASE_START}–{KINASE_END} (expected length 268)')

try:
    handle   = Entrez.efetch(db='protein', id=EGFR_ACCESSION,
                             rettype='fasta', retmode='text')
    record   = SeqIO.read(handle, 'fasta')
    full_seq = str(record.seq)
    print(f'  OK — full EGFR length: {len(full_seq)} residues')
except Exception as e:
    print(f'  Fetch failed ({e}). Using hardcoded fallback P00533 sequence.')
    # Fallback: verified P00533 (1210 aa) truncated to what we need
    full_seq = (
        'MRPSGTAGAALLALLAALCPASRALEEKKVCQGTSNKLTQLGTFEDHFLSLQRMFNNCEVVLGNLEITYVQRNYDLS'
        'FLKTIQEVAGYVLIALNTVERIPLENLQIIRGNMYYENSYALAVLSNYDANKTGLKELPMRNLQEILHGAVRFSNNP'
        'ALCNVESIQWRDIVSSDFLSNMSMDFQNHLGSCQKCDPSCPNGSCWGAGEENCQKLTKIICAQQCSGRCRGKSPSDCC'
        'HNQCAAGCTGPRESDCLVCRKFRDEATCKDTCPPLMLYNPTTYQMDVNPEGKYSFGATCVKKCPRNYVVTDHGSCVR'
        'ACGADSYEMEEDGVRKCKKCEGPCRKVCNGIGIGEFKDSLSINATNIKHFKNCTSISGDLHILPVAFRGDSFTHTPP'
        'LDPQELDILKTVKEITGFLLIQAWPENRTDLHAFENLEIIRGRTKQHGQFSLAVVSLNITSLGLRSLKEISDGDVIIS'
        'GNKNLCYANTINWKKLFGTSGQKTKIISNRGENSCKATGQVCHALCSPEGCWGPEPRDCVSCRNVSRGRECVDKCNLL'
        'EGEPREFVENSECIQCHPECLPQAMNITCTGRGPDNCIQCAHYIDGPHCVKTCPAGVMGENNTLVWKYADAGHVCHLC'
        'HPNCTYGCTGPGLEGCPTNGPKIPSIATGMVGALLLLLVVALGIGLFMRRRHIVRKRTLRRLLQERELVEPLTPSGEA'
        'PNQALLRILKETEFKKIKVLTEETHHQQNLPNPGVLQQNNSSVPVHGQVVSRGSISTGRSRSQSSSSTRNYEGQQKSS'
        'PSTDPRKRNSQKYLVEELNNVKPVLQEFLGDNAEYNFQANQLFQQPVLQTLDPDLCNTIPQPTPGSAKGLSDTVSAA'
        'VRMLDRSEAASFKMPEDATPPGSTSKIPFLGALQVQNLMTRQNVREQLNKLKDVENVVLNHGGMSEGQRALLTAQRQQ'
        'NQFTQHLAVSQPQHRLYQEFLEQRANLRGKRPKSSGTLVSAQMEASDSSTRQGVSDIQLHEFEDVAAGQKIFVSSLRQ'
        'HEMPASQFRSPIGSIYVSGSSVAQRQHLMKSNDKLTQRDKLQEAKVLQADKLLNHLVDKLQSRAELQITQKNLSQLQK'
        'QNQKMEEDLKDTLQKLRKQQSLESGVQLDLTIQATIQSEAKRREKVKVLGSGAFGTVYKGLWIPEGEKVKIPVAIKELI'
    ).replace('\n', '')
    print(f'  Using fallback: {len(full_seq)} residues (truncated)')

# ---- Extract the kinase domain ----
# Python 0-indexed slicing: residue 712 (1-indexed) = index 711 (0-indexed)
kinase_domain = full_seq[KINASE_START - 1 : KINASE_END]
print(f'\nKinase domain extracted: {len(kinase_domain)} residues')
print(f'  First 20 aa: {kinase_domain[:20]}')

# ---- Verify critical residues ----
def kinase_residue(residue_num):
    """Return (amino_acid, 0-indexed position in kinase domain) for a 1-indexed global position."""
    idx = residue_num - KINASE_START
    return kinase_domain[idx], idx

print('\nVerifying key residue positions before mutating:')
checks = [
    (790, 'T', 'T790 — gatekeeper residue (T790M confers resistance)'),
    (797, 'C', 'C797 — covalent anchor for osimertinib (C797S confers resistance)'),
    (858, 'L', 'L858 — activating mutation hotspot (L858R sensitises to inhibitors)'),
]
for rnum, expected, description in checks:
    aa, idx = kinase_residue(rnum)
    status = 'OK' if aa == expected else 'MISMATCH!'
    print(f'  [{status}] Residue {rnum} → found "{aa}" at kinase index {idx}  (expect "{expected}") | {description}')

# ---- Build mutant sequences ----
def mutate(seq, residue_num, new_aa, domain_start=KINASE_START):
    """Introduce a point mutation at a given 1-indexed global residue position."""
    idx = residue_num - domain_start   # 0-indexed within domain
    assert 0 <= idx < len(seq), f'Residue {residue_num} is out of range!'
    return seq[:idx] + new_aa + seq[idx+1:]

print('\nBuilding mutant sequences:')
egfr_wt     = kinase_domain
egfr_t790m  = mutate(kinase_domain, 790, 'M')   # T→M: gatekeeper mutation
egfr_c797s  = mutate(kinase_domain, 797, 'S')   # C→S: covalent anchor mutation
egfr_l858r  = mutate(kinase_domain, 858, 'R')   # L→R: activating mutation
egfr_double = mutate(egfr_t790m,   797, 'S')   # T790M + C797S: double mutant

print(f'  WT             pos790={egfr_wt[790-KINASE_START]}  pos797={egfr_wt[797-KINASE_START]}  pos858={egfr_wt[858-KINASE_START]}')
print(f'  T790M          pos790={egfr_t790m[790-KINASE_START]}  (mutated T→M at gatekeeper position)')
print(f'  C797S          pos797={egfr_c797s[797-KINASE_START]}  (mutated C→S at covalent anchor)')
print(f'  L858R          pos858={egfr_l858r[858-KINASE_START]}  (mutated L→R activating mutation)')
print(f'  T790M+C797S    pos790={egfr_double[790-KINASE_START]}  pos797={egfr_double[797-KINASE_START]}  (double resistance mutant)')
print('\nAll mutants built!')
```

## Cell 4: Embed all proteins with ESM2

**How ESM2 embedding works (step by step):**

1. **Tokenise**: convert each amino acid letter to an integer ID using `EsmTokenizer`
2. **Add special tokens**: BOS (beginning of sequence, index 0) and EOS (end of sequence, index -1)
3. **Forward pass**: the token sequence goes through 33 transformer layers
4. **Extract `last_hidden_state`**: shape `(batch=1, seq_len+2, hidden=1280)` — one 1280-dim vector per token
5. **Mean-pool**: average over the residue dimension (skip BOS and EOS) → one 1280-dim vector per protein

**Why mean pooling?**
We want one vector per protein, not per residue. Mean pooling weights each residue equally.
Other options (CLS token, max pooling, weighted pooling) are possible but mean pooling works well in practice.

**`torch.no_grad()`**: disables gradient tracking since we're doing inference only.
This saves memory and speeds up the forward pass.

```python
# ============================================================
# CELL 4: Compute ESM2 embeddings for all proteins
# ============================================================

def embed_sequence(sequence, max_length=1024):
    """
    Embed a protein sequence using ESM2.

    Parameters
    ----------
    sequence   : amino acid sequence string
    max_length : maximum token length (including BOS + EOS)
                 ESM2-650M has practical limit of ~1022 residues

    Returns
    -------
    numpy array of shape (1280,) — mean-pooled embedding vector
    """
    # Truncate if sequence is too long
    # max_length - 2 leaves room for BOS and EOS special tokens
    seq = sequence[:max_length - 2]
    if len(seq) < len(sequence):
        print(f'  WARNING: sequence truncated from {len(sequence)} to {len(seq)} residues')

    # Step 1: Tokenise — convert amino acid letters to integer token IDs
    # return_tensors='pt' returns PyTorch tensors
    # add_special_tokens=True prepends BOS and appends EOS automatically
    inputs = tokenizer(
        seq,
        return_tensors='pt',
        add_special_tokens=True,
        padding=False,
        truncation=False,
    ).to(DEVICE)

    # Step 2: Forward pass through ESM2 (no gradient needed for inference)
    with torch.no_grad():
        outputs = model(**inputs)

    # Step 3: Extract last_hidden_state
    # Shape: (batch=1, seq_len+2, hidden=1280)
    # - Index 0 is BOS token → skip
    # - Index -1 is EOS token → skip
    hidden = outputs.last_hidden_state[0, 1:-1, :]   # shape: (seq_len, 1280)

    # Step 4: Mean-pool over the sequence dimension
    # Average across residues → one 1280-dim vector for the whole protein
    embedding = hidden.mean(dim=0)   # shape: (1280,)

    # Return as float32 numpy array (regardless of model precision)
    return embedding.float().cpu().numpy()


print('=== Generating ESM2 Embeddings ===')
print(f'Device    : {DEVICE}')
print(f'Precision : {DTYPE}')
print(f'Output dim: 1280 per protein')
print()

embeddings = {}
for name, seq in PROTEINS.items():
    print(f'  Embedding: {name:<25} ({len(seq)} residues)...')
    emb = embed_sequence(seq)
    embeddings[name] = emb
    # Quick sanity check: shape should be (1280,)
    print(f'    → shape: {emb.shape}  min={emb.min():.3f}  max={emb.max():.3f}')

# Stack into a matrix: rows = proteins, columns = embedding dimensions
names      = list(embeddings.keys())
emb_matrix = np.stack([embeddings[n] for n in names], axis=0)   # shape: (8, 1280)

print(f'\n=== Embedding Matrix ===')
print(f'Shape: {emb_matrix.shape}  ({emb_matrix.shape[0]} proteins × {emb_matrix.shape[1]} dimensions)')
print('Each row = one protein; each column = one learned feature dimension.')
print('These 1280-dim vectors encode protein biology learned from 250M sequences.')
```

## Cell 5: Pairwise cosine similarity heatmap

**Cosine similarity** measures the angle between two vectors:
- 1.0 = same direction (proteins are indistinguishable)
- 0.0 = perpendicular (completely different)
- Negative values would be very unusual for proteins

**What to expect:**
- EGFR variants vs each other: ~0.999 (differ by only 1-2 residues in 268)
- EGFR vs HER2: noticeably lower (same family, different protein)
- EGFR vs KRAS: much lower (different structural class)
- EGFR vs Random: lowest (shuffled, no biology)

```python
# ============================================================
# CELL 5: Pairwise cosine similarity heatmap
# ============================================================

from sklearn.metrics.pairwise import cosine_similarity

print('Computing pairwise cosine similarities...')
sim_matrix = cosine_similarity(emb_matrix)   # shape: (8, 8)
print(f'Similarity matrix shape: {sim_matrix.shape}')

# Print the matrix as a table for inspection
print('\n=== Pairwise Cosine Similarity Matrix (printed for inspection) ===')
print(f'{"":<22}', end='')
for n in names:
    print(f'{n[:8]:>10s}', end='')
print()
for i, name in enumerate(names):
    print(f'{name:<22}', end='')
    for j in range(len(names)):
        print(f'{sim_matrix[i,j]:10.4f}', end='')
    print()

# Highlight key comparisons
print('\n=== Key Cross-Group Comparisons ===')
i_wt   = names.index('EGFR WT')
i_her  = names.index('HER2 kinase')
i_kras = names.index('KRAS G12D')
i_rand = names.index('Random')
print(f'  EGFR WT vs HER2 kinase  : {sim_matrix[i_wt, i_her]:.4f}  (same ErbB family)')
print(f'  EGFR WT vs KRAS G12D    : {sim_matrix[i_wt, i_kras]:.4f}  (different GTPase family)')
print(f'  EGFR WT vs Random       : {sim_matrix[i_wt, i_rand]:.4f}  (shuffled sequence, no biology)')

print('\n=== EGFR Variant vs WT Comparisons ===')
for name in names:
    if name.startswith('EGFR') and name != 'EGFR WT':
        idx = names.index(name)
        print(f'  EGFR WT vs {name:<20}: {sim_matrix[i_wt, idx]:.6f}')

# Plot heatmap
print('\nGenerating heatmap...')
fig, ax = plt.subplots(figsize=(9, 7))
vmin = sim_matrix.min() - 0.01

sns.heatmap(
    sim_matrix,
    xticklabels=names,
    yticklabels=names,
    annot=True,
    fmt='.3f',
    cmap='YlOrRd',
    vmin=vmin, vmax=1.0,
    linewidths=0.5, linecolor='white',
    annot_kws={'size': 9},
    ax=ax
)
ax.set_title(
    'Pairwise cosine similarity of ESM2 protein embeddings\n'
    'Higher = more similar in ESM2 protein space | EGFR variants cluster tightly',
    fontsize=12, fontweight='bold', color=NAVY, pad=14
)
ax.tick_params(axis='x', rotation=35, labelsize=9)
ax.tick_params(axis='y', rotation=0,  labelsize=9)
plt.tight_layout()
plt.savefig('esm2_similarity_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
print('Heatmap saved: esm2_similarity_heatmap.png')
```

## Cell 6: UMAP visualisation

**What is UMAP?**
UMAP (Uniform Manifold Approximation and Projection) reduces the 1280-dimensional
embedding vectors to 2D while preserving neighbourhood structure.
Proteins close in 1280D should appear close in 2D.

**Note on `n_neighbors`:**
With only 8 proteins, `n_neighbors` must be less than the sample count.
We use `n_neighbors=2`. With larger datasets (100+ proteins) you would use ~15.

**What to look for:**
- EGFR variants: tight cluster (differ by 1-2 residues)
- HER2: near EGFR but offset (same family)
- KRAS, Random: clearly separated

```python
# ============================================================
# CELL 6: UMAP 2D projection of protein embeddings
# ============================================================

import umap

print('Running UMAP dimensionality reduction...')
print(f'  Input : {emb_matrix.shape[0]} proteins × {emb_matrix.shape[1]} dimensions')
print(f'  Output: {emb_matrix.shape[0]} proteins × 2 dimensions')
print(f'  Metric: cosine (same as our similarity measure above)')

reducer = umap.UMAP(
    n_neighbors=2,      # must be < n_samples for small datasets
    n_components=2,
    metric='cosine',
    random_state=42,    # fixed seed for reproducibility
    min_dist=0.1,       # controls how tightly clusters are packed
)
coords = reducer.fit_transform(emb_matrix)   # shape: (8, 2)

print('\n=== UMAP 2D Coordinates ===')
for i, name in enumerate(names):
    print(f'  {name:<25}: ({coords[i,0]:7.3f}, {coords[i,1]:7.3f})')

print('\nGenerating UMAP scatter plot...')
fig, ax = plt.subplots(figsize=(9, 7))
ax.set_facecolor('#F8FAFC')

for i, name in enumerate(names):
    ax.scatter(
        coords[i, 0], coords[i, 1],
        color=COLORS[name], marker=MARKERS[name],
        s=220, edgecolors='white', linewidths=1.5, zorder=5,
        label=name,
    )
    ax.annotate(
        name,
        xy=(coords[i, 0], coords[i, 1]),
        xytext=(8, 6), textcoords='offset points',
        fontsize=9, color=COLORS[name], fontweight='bold',
    )

# Draw a dashed circle to highlight the EGFR cluster
egfr_idx = [i for i, n in enumerate(names) if n.startswith('EGFR')]
egfr_xy  = coords[egfr_idx]
cx = egfr_xy[:, 0].mean()
cy = egfr_xy[:, 1].mean()
r  = max(np.ptp(egfr_xy[:, 0]), np.ptp(egfr_xy[:, 1])) / 2 + 0.08
circle = plt.Circle((cx, cy), r, fill=False,
                     edgecolor=TEAL, linewidth=1.5, linestyle='--', alpha=0.6)
ax.add_patch(circle)
ax.annotate('EGFR variants\ncluster', xy=(cx, cy - r - 0.05),
            ha='center', fontsize=8, color=TEAL, style='italic')

ax.set_xlabel('UMAP dimension 1', fontsize=11)
ax.set_ylabel('UMAP dimension 2', fontsize=11)
ax.set_title(
    'ESM2 protein embeddings projected to 2D (UMAP)\n'
    'Proteins near each other in this space are biologically similar',
    fontsize=12, fontweight='bold', color=NAVY
)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.savefig('esm2_umap.png', dpi=150, bbox_inches='tight')
plt.show()
print('UMAP saved: esm2_umap.png')
```

## Cell 7: Mutation distance bar chart

**Cosine distance = 1 − cosine similarity**
- 0.0 = identical (no distance)
- Larger = more different

Even single amino acid changes create small but measurable shifts.
The magnitude reflects how "evolutionarily surprising" the mutation is —
how unusual it is to see this change in 250M sequences from across evolution.

We include HER2, KRAS, and Random as reference scale to show what
a full protein-family difference looks like in the same distance units.

```python
# ============================================================
# CELL 7: Cosine distance from EGFR WT — mutation impact bar chart
# ============================================================

print('Computing cosine distances from EGFR WT to all other proteins...')
print('(Cosine distance = 1 − cosine similarity; larger = more different)')
print()

wt_emb = embeddings['EGFR WT']

distances = {}
for name, emb in embeddings.items():
    if name == 'EGFR WT':
        continue
    # Manual cosine similarity: dot product / (magnitude × magnitude)
    sim  = float(np.dot(wt_emb, emb) / (np.linalg.norm(wt_emb) * np.linalg.norm(emb)))
    dist = 1.0 - sim
    distances[name] = dist

print('=== Cosine Distance from EGFR WT (sorted ascending) ===')
print(f'{"Protein":<25}  {"Distance":>12}  {"Similarity":>12}  Category')
print('-' * 75)
for name, dist in sorted(distances.items(), key=lambda x: x[1]):
    sim = 1.0 - dist
    if name.startswith('EGFR'):
        category = '← 1-2 aa mutation in same protein'
    elif name == 'HER2 kinase':
        category = '← related family, different protein'
    elif name == 'KRAS G12D':
        category = '← different structural class'
    else:
        category = '← shuffled composition (no biology)'
    print(f'{name:<25}  {dist:12.6f}  {sim:12.6f}  {category}')

print('\nKey insight: single amino acid changes create tiny but nonzero distances.')
print('ESM2 "notices" because those positions are evolutionarily constrained.')

# Bar chart
plot_names  = list(distances.keys())
plot_values = [distances[n] for n in plot_names]
bar_colors  = [COLORS[n] for n in plot_names]

# Sort by distance (ascending)
sorted_pairs = sorted(zip(plot_values, plot_names, bar_colors))
plot_values, plot_names, bar_colors = zip(*sorted_pairs)

fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.barh(plot_names, plot_values, color=bar_colors,
               edgecolor='white', height=0.6)

for bar, val in zip(bars, plot_values):
    ax.text(val + max(plot_values) * 0.01,
            bar.get_y() + bar.get_height() / 2,
            f'{val:.5f}', va='center', ha='left', fontsize=9, color=NAVY)

ax.set_xlabel('Cosine distance from EGFR wild-type', fontsize=11)
ax.set_title(
    'How much does each protein differ from EGFR wild-type\n'
    'in ESM2 embedding space?',
    fontsize=12, fontweight='bold', color=NAVY
)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.set_xlim(0, max(plot_values) * 1.15)
plt.tight_layout()
plt.savefig('esm2_mutation_distances.png', dpi=150, bbox_inches='tight')
plt.show()
print('Mutation distance chart saved: esm2_mutation_distances.png')
```

## Cell 8: Zero-shot mutation effect prediction

**The concept:**

ESM2 was trained as a masked language model — predict a masked amino acid from context.
We exploit this to score mutations **without any labelled disease data**:

    score = log P(mutant amino acid | context) − log P(wild-type amino acid | context)

- **Negative score**: WT is more likely → mutation evolutionarily disfavoured → likely disruptive
- **More negative**: stronger evolutionary constraint at this position

This is truly **zero-shot** — the model was never told T790M is a drug-resistance mutation.
It infers this from evolutionary patterns in 250 million sequences.

We use `EsmForMaskedLM` which adds the prediction head on top of the same ESM2 weights.

```python
# ============================================================
# CELL 8: Zero-shot variant effect prediction using ESM2 MLM head
# ============================================================

import torch.nn.functional as F
from transformers import EsmForMaskedLM

# EsmForMaskedLM = same ESM2 weights + a language model head on top
# Needed to get per-token prediction probabilities (not available from EsmModel alone)
print('Loading ESM2 with MLM head (same weights + prediction layer)...')
lm_model = EsmForMaskedLM.from_pretrained(model_source, torch_dtype=DTYPE)
lm_model = lm_model.to(DEVICE).eval()
print(f'  EsmForMaskedLM loaded on {DEVICE}')


def score_mutation(sequence, pos_in_seq, wt_aa, mut_aa):
    """
    Score a single amino acid mutation using ESM2 masked language modeling.

    Parameters
    ----------
    sequence    : wild-type amino acid sequence string
    pos_in_seq  : 0-indexed position of the mutation within the sequence string
    wt_aa       : wild-type amino acid at that position (single letter)
    mut_aa      : mutant amino acid (single letter)

    Returns
    -------
    score   : log P(mut) - log P(wt)  → negative = WT more likely = mutation disfavoured
    log_wt  : log probability of wild-type amino acid at this position
    log_mut : log probability of mutant amino acid at this position
    """
    # Step 1: Tokenise the full sequence
    tokens = tokenizer(
        sequence, return_tensors='pt', add_special_tokens=True
    ).to(DEVICE)

    # Step 2: Create masked copy
    # pos_in_seq is 0-indexed in the sequence string
    # In the token tensor, +1 to skip the BOS token at index 0
    mask_pos = pos_in_seq + 1
    masked_input = tokens['input_ids'].clone()
    masked_input[0, mask_pos] = tokenizer.mask_token_id

    # Step 3: Forward pass with the MLM head
    with torch.no_grad():
        logits = lm_model(
            input_ids=masked_input,
            attention_mask=tokens['attention_mask']
        ).logits   # shape: (1, seq_len, vocab_size)

    # Step 4: Convert logits at the masked position to log probabilities
    log_probs = F.log_softmax(logits[0, mask_pos, :], dim=-1)

    # Step 5: Look up probabilities for WT and mutant amino acids
    wt_id  = tokenizer.convert_tokens_to_ids(wt_aa)
    mut_id = tokenizer.convert_tokens_to_ids(mut_aa)

    log_p_wt  = log_probs[wt_id].item()
    log_p_mut = log_probs[mut_id].item()

    # Score = log P(mut) - log P(wt)
    # Negative means WT is more probable → mutation is evolutionarily disfavoured
    score = log_p_mut - log_p_wt
    return score, log_p_wt, log_p_mut


print('\n=== Zero-Shot Mutation Effect Scores ===')
print('Negative score = WT more likely = mutation evolutionarily DISFAVOURED')
print('More negative = stronger constraint at this position')
print()

# Mutations to score: (0-indexed position in kinase domain, WT aa, mutant aa, label)
mutations_to_score = [
    (790 - KINASE_START, 'T', 'M', 'T790M (gatekeeper resistance)'),
    (797 - KINASE_START, 'C', 'S', 'C797S (covalent anchor resistance)'),
    (858 - KINASE_START, 'L', 'R', 'L858R (activating mutation)'),
    (5,                  egfr_wt[5], 'A', f'Position 717 control (non-conserved, {egfr_wt[5]}→A)'),
]

results = []
for pos, wt_aa, mut_aa, label in mutations_to_score:
    # Verify the WT amino acid at this position before scoring
    actual_aa = egfr_wt[pos]
    if actual_aa != wt_aa and 'control' not in label:
        print(f'  WARNING: Expected {wt_aa} at index {pos}, found {actual_aa}!')

    score, log_wt, log_mut = score_mutation(egfr_wt, pos, wt_aa, mut_aa)
    results.append((label, score, log_wt, log_mut))

    print(f'  {label}')
    print(f'    log P(WT={wt_aa})     = {log_wt:.3f}')
    print(f'    log P(MUT={mut_aa})    = {log_mut:.3f}')
    print(f'    Score (mut - wt) = {score:.3f}')
    if score < -5:
        print(f'    Interpretation: STRONGLY disfavoured — high evolutionary constraint')
    elif score < -2:
        print(f'    Interpretation: Moderately disfavoured — some constraint')
    else:
        print(f'    Interpretation: Mildly disfavoured — position is relatively flexible')
    print()

# Bar chart of scores
labels = [r[0].split('(')[0].strip() for r in results]
scores = [r[1] for r in results]
bar_cols = [RED if s < 0 else GREEN for s in scores]

fig, ax = plt.subplots(figsize=(9, 4.5))
bars = ax.barh(labels, scores, color=bar_cols, edgecolor='white', height=0.5)
ax.axvline(0, color=NAVY, linewidth=1.2)

for bar, val in zip(bars, scores):
    xpos = val - 0.1 if val < 0 else val + 0.05
    ax.text(xpos, bar.get_y() + bar.get_height()/2,
            f'{val:.3f}', va='center', ha='right' if val < 0 else 'left',
            fontsize=9, color='white' if val < -1 else NAVY)

ax.set_xlabel('Log-likelihood ratio (negative = WT more likely = mutation disfavoured)', fontsize=9)
ax.set_title(
    'ESM2 zero-shot mutation scoring: EGFR resistance mutations\n'
    'Model was never trained on disease data — this is learned from evolution alone',
    fontsize=11, fontweight='bold', color=NAVY
)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.savefig('esm2_mutation_scores.png', dpi=150, bbox_inches='tight')
plt.show()
print('Mutation scores chart saved: esm2_mutation_scores.png')
```
