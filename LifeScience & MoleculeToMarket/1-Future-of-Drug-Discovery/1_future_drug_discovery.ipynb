### What this notebook demonstrates

This notebook shows how **ESM2** (a protein language model from Meta AI) can represent
protein sequences as mathematical vectors called **embeddings**. We use EGFR — a cancer
drug target — and its known resistance mutations to show that ESM2 understands biology
without ever being told what the mutations mean.

- What a protein language model is and why it matters
- How to get protein embeddings from ESM2
- How to measure similarity between proteins using cosine similarity
- How to visualise protein space using UMAP

### EGFR mutations we study
| Mutation | Position | Effect |
|----------|----------|--------|
| T790M | 790 | Gatekeeper resistance — blocks 1st/2nd gen inhibitors |
| C797S | 797 | Covalent anchor resistance — blocks osimertinib |
| L858R | 858 | Activating mutation — makes cancer sensitive to inhibitors |

1. Load ESM2 model
2. Fetch EGFR kinase domain sequence from UniProt, build mutants
3. Define protein panel (EGFR variants + HER2 + KRAS + Random)

5. Pairwise cosine similarity heatmap
6. UMAP visualisation
7. Zero-shot mutation scoring

## Cell 1: Install libraries and load ESM2

**What is ESM2?**
ESM2 is a transformer model trained on 250 million protein sequences. It learns to predict
masked amino acids — just like BERT for text. The model learns protein structure, function,
and evolution implicitly from sequence patterns alone.

**What we need:**
- `transformers`: HuggingFace library to load ESM2
- `biopython`: to fetch EGFR sequence from UniProt
- `umap-learn`: for 2D visualisation of embeddings
- `torch`: PyTorch for running the model

**GPU vs CPU:** ESM2 runs faster on GPU (float16). On CPU it uses float32 and is slower
but still works. We auto-detect which is available.

## Cell 2: Fetch EGFR kinase domain and build mutants

**Why the kinase domain?**
The full EGFR protein is 1210 amino acids. The kinase domain (residues 712–979) is where
all the drug-binding action happens. We use just this 268 aa domain to keep embeddings
efficient and focused on the relevant region.

**Mutation indexing:**
UniProt uses 1-based residue numbering. When slicing the kinase domain:
- kinase domain starts at residue 712 (1-indexed)
- Residue 790 → index (790 - 712) = 78 within the kinase domain string
- We verify each position has the expected wild-type amino acid before mutating

```python
# ============================================================
# CELL 2: Fetch EGFR sequence from UniProt and build mutants
# ============================================================

from Bio import Entrez, SeqIO

# NCBI requires an email for Entrez queries (not validated, just for contact)
Entrez.email = 'your@email.com'

# EGFR UniProt accession and kinase domain boundaries
EGFR_ACCESSION = 'P00533'   # human EGFR (ErbB-1) on UniProt
KINASE_START   = 712        # 1-indexed, inclusive
KINASE_END     = 979        # 1-indexed, inclusive

print(f'Fetching EGFR protein sequence from UniProt...')
print(f'  Accession : {EGFR_ACCESSION}')
print(f'  Kinase domain: residues {KINASE_START}–{KINASE_END}')

# ---- Fetch from UniProt via NCBI Entrez ----
try:
    handle   = Entrez.efetch(db='protein', id=EGFR_ACCESSION,
                             rettype='fasta', retmode='text')
    record   = SeqIO.read(handle, 'fasta')
    full_seq = str(record.seq)
    print(f'\nSequence fetched from UniProt!')
    print(f'  Full EGFR length: {len(full_seq)} residues')
except Exception as e:
    print(f'  Fetch failed ({e}). Using hardcoded fallback sequence.')
    # Fallback: truncated P00533 sequence (first part only)
    full_seq = (
        'MRPSGTAGAALLALLAALCPASRALEEKKVCQGTSNKLTQLGTFEDHFLSLQRMFNNCEVVLGNLEITYVQRNYDLS'
        'FLKTIQEVAGYVLIALNTVERIPLENLQIIRGNMYYENSYALAVLSNYDANKTGLKELPMRNLQEILHGAVRFSNNP'
        'ALCNVESIQWRDIVSSDFLSNMSMDFQNHLGSCQKCDPSCPNGSCWGAGEENCQKLTKIICAQQCSGRCRGKSPSDCC'
        'HNQCAAGCTGPRESDCLVCRKFRDEATCKDTCPPLMLYNPTTYQMDVNPEGKYSFGATCVKKCPRNYVVTDHGSCVR'
        'ACGADSYEMEEDGVRKCKKCEGPCRKVCNGIGIGEFKDSLSINATNIKHFKNCTSISGDLHILPVAFRGDSFTHTPP'
        'LDPQELDILKTVKEITGFLLIQAWPENRTDLHAFENLEIIRGRTKQHGQFSLAVVSLNITSLGLRSLKEISDGDVIIS'
        'GNKNLCYANTINWKKLFGTSGQKTKIISNRGENSCKATGQVCHALCSPEGCWGPEPRDCVSCRNVSRGRECVDKCNLL'
        'EGEPREFVENSECIQCHPECLPQAMNITCTGRGPDNCIQCAHYIDGPHCVKTCPAGVMGENNTLVWKYADAGHVCHLC'
        'HPNCTYGCTGPGLEGCPTNGPKIPSIATGMVGALLLLLVVALGIGLFMRRRHIVRKRTLRRLLQERELVEPLTPSGEA'
        'PNQALLRILKETEFKKIKVLTEETHHQQNLPNPGVLQQNNSSVPVHGQVVSRGSISTGRSRSQSSSSTRNYEGQQKSS'
        'PSTDPRKRNSQKYLVEELNNVKPVLQEFLGDNAEYNFQANQLFQQPVLQTLDPDLCNTIPQPTPGSAKGLSDTVSAA'
        'VRMLDRSEAASFKMPEDATPPGSTSKIPFLGALQVQNLMTRQNVREQLNKLKDVENVVLNHGGMSEGQRALLTAQRQQ'
        'NQFTQHLAVSQPQHRLYQEFLEQRANLRGKRPKSSGTLVSAQMEASDSSTRQGVSDIQLHEFEDVAAGQKIFVSSLRQ'
        'HEMPASQFRSPIGSIYVSGSSVAQRQHLMKSNDKLTQRDKLQEAKVLQADKLLNHLVDKLQSRAELQITQKNLSQLQK'
        'QNQKMEEDLKDTLQKLRKQQSLESGVQLDLTIQATIQSEAKRREKVKVLGSGAFGTVYKGLWIPEGEKVKIPVAIKELI'
    ).replace('\n', '')
    print(f'  Fallback sequence: {len(full_seq)} residues (truncated)')

# ---- Extract the kinase domain ----
# Python slicing is 0-indexed, so residue 712 (1-indexed) = index 711 (0-indexed)
kinase_domain = full_seq[KINASE_START - 1 : KINASE_END]
print(f'\nKinase domain extracted: {len(kinase_domain)} residues')

# ---- Verify critical residues before mutating ----
print('\nVerifying key residues at correct positions:')
checks = [
    (790, 'T', 'T790 — gatekeeper residue (T790M confers resistance)'),
    (797, 'C', 'C797 — covalent anchor for osimertinib (C797S confers resistance)'),
    (858, 'L', 'L858 — activating mutation hotspot (L858R sensitises to inhibitors)'),
]
all_ok = True
for res_num, expected_aa, description in checks:
    idx    = res_num - KINASE_START   # 0-indexed within kinase domain
    actual = kinase_domain[idx]
    status = 'OK' if actual == expected_aa else 'MISMATCH'
    print(f'  [{status}] Residue {res_num} → found "{actual}" (expected "{expected_aa}") | {description}')
    if actual != expected_aa:
        all_ok = False

if all_ok:
    print('\nAll residue checks passed! Proceeding to build mutants.')
else:
    print('\nWARNING: Some residues did not match. Check your sequence source.')

# ---- Build mutant sequences ----
def mutate(seq, residue_num, new_aa, domain_start=KINASE_START):
    """Introduce a point mutation at a specific 1-indexed position."""
    idx = residue_num - domain_start  # convert to 0-indexed
    assert 0 <= idx < len(seq), f'Residue {residue_num} is outside the domain range!'
    mutated = seq[:idx] + new_aa + seq[idx+1:]
    return mutated

print('\nBuilding mutant sequences:')
egfr_wt    = kinase_domain
egfr_t790m = mutate(kinase_domain, 790, 'M')   # T→M: gatekeeper mutation
egfr_c797s = mutate(kinase_domain, 797, 'S')   # C→S: covalent anchor mutation
egfr_l858r = mutate(kinase_domain, 858, 'R')   # L→R: activating mutation
egfr_double = mutate(egfr_t790m, 797, 'S')     # T790M + C797S double mutant

# Verify mutations were introduced correctly
print(f'  EGFR WT         pos790={egfr_wt[790-KINASE_START]}  pos797={egfr_wt[797-KINASE_START]}  pos858={egfr_wt[858-KINASE_START]}')
print(f'  EGFR T790M      pos790={egfr_t790m[790-KINASE_START]}  (mutated T→M)')
print(f'  EGFR C797S      pos797={egfr_c797s[797-KINASE_START]}  (mutated C→S)')
print(f'  EGFR L858R      pos858={egfr_l858r[858-KINASE_START]}  (mutated L→R)')
print(f'  EGFR T790M+C797S pos790={egfr_double[790-KINASE_START]}  pos797={egfr_double[797-KINASE_START]}  (double mutant)')
print('\nAll mutants built successfully!')
```

## Cell 4: Generate ESM2 embeddings

**How ESM2 embedding works:**
1. Tokenise: convert each amino acid letter to an integer ID
2. Add special tokens: BOS (beginning of sequence) and EOS (end of sequence)
3. Forward pass through 33 transformer layers
4. Extract `last_hidden_state`: shape `(1, seq_len+2, 1280)` — one 1280-dim vector per token
5. Mean-pool over sequence dimension (excluding BOS/EOS) → one 1280-dim vector per protein

**Why mean pooling?**
We want one number per protein, not one per residue. Mean pooling gives each residue
equal weight. This is not the only choice — other options include CLS token, max pooling,
or weighted pooling — but mean pooling works well in practice.

**`torch.no_grad()`:** We're doing inference only, not training. This disables gradient
tracking, which saves memory and speeds up computation.

```python
# ============================================================
# CELL 4: Compute ESM2 embeddings for all proteins
# ============================================================

def embed_sequence(sequence, max_length=1024):
    """
    Embed a protein sequence using ESM2.

    Parameters
    ----------
    sequence   : amino acid sequence (single-letter codes)
    max_length : maximum number of tokens (including BOS + EOS)
                 ESM2-650M has practical limit of ~1022 residues

    Returns
    -------
    numpy array of shape (1280,) — mean-pooled embedding vector
    """
    # Truncate if sequence is longer than the model's context window
    # (max_length - 2 leaves room for BOS and EOS special tokens)
    seq = sequence[:max_length - 2]
    if len(seq) < len(sequence):
        print(f'    WARNING: Sequence truncated from {len(sequence)} to {len(seq)} residues')

    # Step 1: Tokenise — convert amino acid letters to integer IDs
    inputs = tokenizer(
        seq,
        return_tensors='pt',      # return PyTorch tensors
        add_special_tokens=True,  # add BOS and EOS tokens
        padding=False,
        truncation=False,
    ).to(DEVICE)

    # Step 2: Forward pass (no gradient computation needed for inference)
    with torch.no_grad():
        outputs = model(**inputs)

    # Step 3: Extract residue embeddings
    # last_hidden_state shape: (batch=1, seq_len+2, hidden=1280)
    # Index 0 = BOS token, index -1 = EOS token → skip these
    hidden = outputs.last_hidden_state[0, 1:-1, :]  # shape: (seq_len, 1280)

    # Step 4: Mean pool across the sequence dimension
    embedding = hidden.mean(dim=0)  # shape: (1280,)

    # Return as float32 numpy array (regardless of model precision)
    return embedding.float().cpu().numpy()


print('=== Generating ESM2 Embeddings ===')
print(f'Device: {DEVICE} | Precision: {DTYPE}')
print(f'Embedding dimension: 1280 (output of ESM2-650M)')
print()

embeddings = {}
for name, seq in PROTEINS.items():
    print(f'  Embedding: {name:<25} ({len(seq)} residues)...')
    emb = embed_sequence(seq)
    embeddings[name] = emb
    # Quick sanity check: embedding should be a 1D vector of length 1280
    print(f'    → embedding shape: {emb.shape}  min={emb.min():.3f}  max={emb.max():.3f}')

# Stack into a matrix for downstream computation
names      = list(embeddings.keys())
emb_matrix = np.stack([embeddings[n] for n in names], axis=0)  # shape: (8, 1280)

print(f'\n=== Embedding Matrix ===')
print(f'Shape: {emb_matrix.shape}  ({emb_matrix.shape[0]} proteins × {emb_matrix.shape[1]} dimensions)')
print('Each row is one protein; each column is one learned feature dimension.')
print('These 1280-dimensional vectors encode protein biology as learned from 250M sequences.')
```

## Cell 5: Pairwise cosine similarity heatmap

**Cosine similarity** measures the angle between two vectors:
- 1.0 = identical direction (proteins are indistinguishable)
- 0.0 = perpendicular (completely different)
- -1.0 = opposite (would be very unusual for proteins)

**What to look for in the heatmap:**
- EGFR variants should form a bright cluster (very high similarity ~0.999)
- HER2 vs EGFR should be clearly lower (same family but different protein)
- KRAS vs EGFR should be much lower (different family)
- Random vs anything should be lowest (no biological relationship)

```python
# ============================================================
# CELL 5: Pairwise cosine similarity heatmap
# ============================================================

from sklearn.metrics.pairwise import cosine_similarity

print('Computing pairwise cosine similarities between all protein embeddings...')
sim_matrix = cosine_similarity(emb_matrix)  # shape: (8, 8)
print(f'Similarity matrix shape: {sim_matrix.shape}')

# ---- Print the similarity matrix as a table ----
print('\n=== Pairwise Cosine Similarity Matrix ===')
header = ''.join(f'{n[:8]:>10s}' for n in names)
print(f'{"":<22}' + header)
for i, name in enumerate(names):
    row = ''.join(f'{sim_matrix[i,j]:10.4f}' for j in range(len(names)))
    print(f'{name:<22}' + row)

# ---- Interpretation guide ----
print('\n=== Key Observations ===')
egfr_names = [n for n in names if n.startswith('EGFR')]
for n1 in egfr_names:
    for n2 in egfr_names:
        if n1 < n2:
            i1, i2 = names.index(n1), names.index(n2)
            print(f'  {n1} vs {n2}: {sim_matrix[i1,i2]:.4f} (differ by 1-2 aa)')

i_wt  = names.index('EGFR WT')
i_her = names.index('HER2 kinase')
i_kras= names.index('KRAS G12D')
i_rand= names.index('Random')
print(f'\n  EGFR WT vs HER2    : {sim_matrix[i_wt, i_her]:.4f} (same ErbB family)')
print(f'  EGFR WT vs KRAS    : {sim_matrix[i_wt, i_kras]:.4f} (different family)')
print(f'  EGFR WT vs Random  : {sim_matrix[i_wt, i_rand]:.4f} (shuffled sequence)')
print(f'\nNote: All similarities are > 0 because all proteins share the same amino acid alphabet.')
print('What matters is the RELATIVE differences between groups.')

# ---- Plot heatmap ----
print('\nGenerating heatmap...')
fig, ax = plt.subplots(figsize=(9, 7))

vmin = sim_matrix.min() - 0.01
vmax = 1.0

sns.heatmap(
    sim_matrix,
    xticklabels=names,
    yticklabels=names,
    annot=True,
    fmt='.3f',
    cmap='YlOrRd',
    vmin=vmin,
    vmax=vmax,
    linewidths=0.5,
    linecolor='white',
    annot_kws={'size': 9},
    ax=ax
)
ax.set_title(
    'Pairwise Cosine Similarity of ESM2 Protein Embeddings\n'
    'Higher = more similar in ESM2 protein space | EGFR variants cluster tightly',
    fontsize=12, fontweight='bold', color=NAVY, pad=14
)
ax.tick_params(axis='x', rotation=35, labelsize=9)
ax.tick_params(axis='y', rotation=0, labelsize=9)
plt.tight_layout()
plt.savefig('esm2_similarity_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
print('Heatmap saved as esm2_similarity_heatmap.png')
```

## Cell 6: UMAP visualisation of embedding space

**What is UMAP?**
UMAP (Uniform Manifold Approximation and Projection) reduces the 1280-dimensional embedding
vectors to 2D while preserving the neighbourhood structure — proteins that are close in
1280D should be close in 2D.

**Note on small datasets:**
With only 8 proteins, `n_neighbors` must be less than the number of samples (8).
We use `n_neighbors=2`. With more proteins, you would use larger values (e.g., 15).

**What to look for:**
- EGFR variants should cluster together (tight blob)
- HER2 should be near EGFR but offset
- KRAS and Random should be far from the EGFR cluster

```python
# ============================================================
# CELL 6: UMAP 2D projection of protein embeddings
# ============================================================

import umap

print('Running UMAP dimensionality reduction...')
print(f'  Input: {emb_matrix.shape[0]} proteins × {emb_matrix.shape[1]} dimensions')
print(f'  Output: {emb_matrix.shape[0]} proteins × 2 dimensions')
print(f'  Metric: cosine (same as our similarity measure)')
print(f'  n_neighbors=2 (minimum for 8 proteins, would use ~15 for larger datasets)')

reducer = umap.UMAP(
    n_neighbors=2,
    n_components=2,
    metric='cosine',
    random_state=42,     # fixed seed for reproducibility
    min_dist=0.1,        # controls how tightly points cluster
)
coords = reducer.fit_transform(emb_matrix)  # shape: (8, 2)

print('\n=== UMAP 2D Coordinates ===')
for i, name in enumerate(names):
    print(f'  {name:<25}: ({coords[i,0]:7.3f}, {coords[i,1]:7.3f})')

# ---- Plot ----
print('\nGenerating UMAP scatter plot...')
fig, ax = plt.subplots(figsize=(9, 7))
ax.set_facecolor('#F8FAFC')

for i, name in enumerate(names):
    ax.scatter(
        coords[i, 0], coords[i, 1],
        color=COLORS[name],
        marker=MARKERS[name],
        s=220,
        edgecolors='white',
        linewidths=1.5,
        zorder=5,
        label=name,
    )
    ax.annotate(
        name,
        xy=(coords[i, 0], coords[i, 1]),
        xytext=(8, 6),
        textcoords='offset points',
        fontsize=9,
        color=COLORS[name],
        fontweight='bold',
    )

# Draw circle highlighting EGFR cluster
egfr_idx = [i for i, n in enumerate(names) if n.startswith('EGFR')]
egfr_xy  = coords[egfr_idx]
cx = egfr_xy[:, 0].mean()
cy = egfr_xy[:, 1].mean()
r  = max(np.ptp(egfr_xy[:, 0]), np.ptp(egfr_xy[:, 1])) / 2 + 0.08
circle = plt.Circle((cx, cy), r, fill=False,
                     edgecolor=TEAL, linewidth=1.5, linestyle='--', alpha=0.6)
ax.add_patch(circle)
ax.annotate('EGFR variants\ncluster', xy=(cx, cy - r - 0.05),
            ha='center', fontsize=8, color=TEAL, style='italic')

ax.set_xlabel('UMAP dimension 1', fontsize=11)
ax.set_ylabel('UMAP dimension 2', fontsize=11)
ax.set_title(
    'ESM2 Protein Embeddings: UMAP 2D Projection\n'
    'Proteins near each other in this space are biologically similar',
    fontsize=12, fontweight='bold', color=NAVY
)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.savefig('esm2_umap.png', dpi=150, bbox_inches='tight')
plt.show()
print('UMAP plot saved as esm2_umap.png')
```

## Cell 7: Mutation distance bar chart

**Cosine distance** = 1 − cosine similarity
- 0.0 = identical (no distance)
- Larger values = more different

**What can we learn from mutation distances?**
Even single amino acid changes create small but measurable shifts in embedding space.
The magnitude of the shift reflects how 'evolutionarily surprising' the mutation is —
how unusual it is to see this change in the protein's evolutionary history.

We also include HER2, KRAS, and Random for reference scale — these show what a full
protein family difference looks like in the same distance units.

```python
# ============================================================
# CELL 7: Cosine distance from EGFR WT — mutation impact bar chart
# ============================================================

print('Computing cosine distances from EGFR WT to all other proteins...')
print('(Distance = 1 − similarity; larger = more different from WT)')
print()

wt_emb = embeddings['EGFR WT']

distances = {}
for name, emb in embeddings.items():
    if name == 'EGFR WT':
        continue
    # Manual cosine similarity calculation
    sim  = float(np.dot(wt_emb, emb) / (np.linalg.norm(wt_emb) * np.linalg.norm(emb)))
    dist = 1.0 - sim
    distances[name] = dist

# Print sorted results
print('=== Cosine Distance from EGFR WT (sorted ascending) ===')
print(f'{"Protein":<25} {"Distance":>12} {"Similarity":>12}')
print('-' * 55)
for name, dist in sorted(distances.items(), key=lambda x: x[1]):
    sim = 1.0 - dist
    category = ''
    if name.startswith('EGFR'):
        category = '← 1-2 aa mutation'
    elif name == 'HER2 kinase':
        category = '← related family'
    elif name == 'KRAS G12D':
        category = '← different family'
    else:
        category = '← shuffled control'
    print(f'{name:<25} {dist:12.6f} {sim:12.6f}  {category}')

print('\nKey insight: Even 1-2 amino acid changes create tiny but nonzero distances.')
print('ESM2 "notices" these changes because they affect conserved functional positions.')

# ---- Bar chart ----
plot_names  = list(distances.keys())
plot_values = [distances[n] for n in plot_names]
bar_colors  = [COLORS[n] for n in plot_names]

# Sort by distance
sorted_pairs = sorted(zip(plot_values, plot_names, bar_colors))
plot_values, plot_names, bar_colors = zip(*sorted_pairs)

fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.barh(plot_names, plot_values, color=bar_colors, edgecolor='white', height=0.6)

# Value labels on bars
for bar, val in zip(bars, plot_values):
    ax.text(val + max(plot_values) * 0.01, bar.get_y() + bar.get_height() / 2,
            f'{val:.5f}', va='center', ha='left', fontsize=9, color=NAVY)

ax.set_xlabel('Cosine Distance from EGFR Wild-Type', fontsize=11)
ax.set_title(
    'How Much Does Each Protein Differ from EGFR Wild-Type\n'
    'in ESM2 Embedding Space?',
    fontsize=12, fontweight='bold', color=NAVY
)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.set_xlim(0, max(plot_values) * 1.15)
plt.tight_layout()
plt.savefig('esm2_mutation_distances.png', dpi=150, bbox_inches='tight')
plt.show()
print('Distance bar chart saved as esm2_mutation_distances.png')
```

## Cell 8: Zero-shot mutation effect prediction

**The concept:**
ESM2 was trained as a masked language model — it predicts masked amino acids from context.
We can exploit this to score mutations WITHOUT any labelled disease data.

**Log-likelihood ratio score:**

    score = log P(mutant amino acid | context) − log P(wild-type amino acid | context)

- **Negative score**: WT is more likely → mutation is evolutionarily disfavored → likely harmful
- **More negative**: stronger evolutionary constraint at this position → more impactful mutation

**This is truly zero-shot** — the model was never told T790M is a drug resistance mutation!
It infers this from evolutionary patterns in 250 million sequences.

We score 3 real EGFR mutations + 1 control position to validate the approach.

```python
# ============================================================
# CELL 8: Zero-shot variant effect prediction using ESM2 MLM head
# ============================================================

import torch.nn.functional as F
from transformers import EsmForMaskedLM

# We need the version of ESM2 with the language model head (for predicting amino acids).
# This uses the SAME weights as before, just adds the prediction layer on top.
print('Loading ESM2 with MLM head for mutation scoring...')
print('This reuses the same pre-trained weights + adds a prediction layer.')

lm_model = EsmForMaskedLM.from_pretrained(model_source, torch_dtype=DTYPE)
lm_model = lm_model.to(DEVICE).eval()
print(f'EsmForMaskedLM loaded on {DEVICE}')


def score_mutation(sequence, pos_in_seq, wt_aa, mut_aa):
    """
    Score a single amino acid mutation using ESM2 masked language modeling.

    Parameters
    ----------
    sequence    : amino acid sequence string (wild-type)
    pos_in_seq  : 0-indexed position of the mutation within sequence string
    wt_aa       : wild-type amino acid at that position (single letter)
    mut_aa      : mutant amino acid (single letter)

    Returns
    -------
    score   : log P(mut) - log P(wt)  → negative = WT more likely = mutation disfavored
    log_wt  : log probability of wild-type amino acid
    log_mut : log probability of mutant amino acid
    """
    # Step 1: Tokenise the full sequence
    tokens = tokenizer(sequence, return_tensors='pt',
                       add_special_tokens=True).to(DEVICE)

    # Step 2: Create masked copy (pos_in_seq + 1 to account for BOS token)
    mask_pos = pos_in_seq + 1
    masked_input = tokens['input_ids'].clone()
    masked_input[0, mask_pos] = tokenizer.mask_token_id

    # Step 3: Forward pass to get predicted probabilities at masked position
    with torch.no_grad():
        logits = lm_model(
            input_ids=masked_input,
            attention_mask=tokens['attention_mask']
        ).logits  # shape: (1, seq_len, vocab_size)

    # Step 4: Convert logits to log-probabilities at the masked position
    log_probs = F.log_softmax(logits[0, mask_pos, :], dim=-1)

    # Step 5: Look up probabilities for wild-type and mutant amino acids
    wt_id  = tokenizer.convert_tokens_to_ids(wt_aa)
    mut_id = tokenizer.convert_tokens_to_ids(mut_aa)

    log_p_wt  = log_probs[wt_id].item()
    log_p_mut = log_probs[mut_id].item()

    # Score = log P(mut) - log P(wt)
    # Negative → WT more probable → mutation is evolutionarily disfavored
    score = log_p_mut - log_p_wt
    return score, log_p_wt, log_p_mut


print('\n=== Zero-Shot Mutation Effect Scores ===')
print('Negative score = WT more likely = mutation is evolutionarily DISFAVORED')
print('More negative = stronger evolutionary constraint at this position')
print()

# Define mutations to score: (1-indexed residue number, WT aa, mutant aa, label)
mutations = [
    (790, 'T', 'M', 'T790M — gatekeeper resistance mutation'),
    (797, 'C', 'S', 'C797S — covalent anchor resistance mutation'),
    (858, 'L', 'R', 'L858R — activating mutation'),
    (  5, egfr_wt[5-0], 'A', f'Position 5 control — {egfr_wt[4]}→A (non-conserved, should be mild)'),
]

results = []
for res_num, wt_aa, mut_aa, label in mutations:
    pos = res_num - KINASE_START   # 0-indexed within kinase domain
    # Verify WT amino acid before scoring
    actual_aa = egfr_wt[pos]
    if actual_aa != wt_aa and label != mutations[-1][3]:  # skip for control
        print(f'  WARNING: Expected {wt_aa} at position {res_num}, found {actual_aa}')

    score, log_wt, log_mut = score_mutation(egfr_wt, pos, wt_aa, mut_aa)
    results.append((label, score, log_wt, log_mut))

    print(f'  {label}')
    print(f'    log P(WT={wt_aa})     = {log_wt:.3f}')
    print(f'    log P(MUT={mut_aa})    = {log_mut:.3f}')
    print(f'    Score (mut - wt) = {score:.3f}')
    if score < -5:
        print(f'    Interpretation: STRONGLY disfavored — high evolutionary constraint here')
    elif score < -2:
        print(f'    Interpretation: Moderately disfavored — some evolutionary constraint')
    else:
        print(f'    Interpretation: Mildly disfavored — position is relatively flexible')
    print()

# ---- Bar chart ----
labels = [r[0].split(' — ')[0] for r in results]
scores = [r[1] for r in results]
bar_cols = [RED if s < 0 else GREEN for s in scores]

fig, ax = plt.subplots(figsize=(9, 4.5))
bars = ax.barh(labels, scores, color=bar_cols, edgecolor='white', height=0.5)
ax.axvline(0, color=NAVY, linewidth=1.2)

for bar, val in zip(bars, scores):
    xpos = val - 0.1 if val < 0 else val + 0.05
    ax.text(xpos, bar.get_y() + bar.get_height()/2,
            f'{val:.3f}', va='center', ha='right' if val < 0 else 'left',
            fontsize=9, color='white' if val < -1 else NAVY)

ax.set_xlabel('Log-Likelihood Ratio  (negative = WT more likely = mutation evolutionarily disfavored)',
              fontsize=9)
ax.set_title(
    'ESM2 Zero-Shot Mutation Scoring: EGFR Resistance Mutations\n'
    'Model was never trained on disease data — this is learned from evolution alone',
    fontsize=11, fontweight='bold', color=NAVY
)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.savefig('esm2_mutation_scores.png', dpi=150, bbox_inches='tight')
plt.show()
print('Mutation score chart saved as esm2_mutation_scores.png')
```
