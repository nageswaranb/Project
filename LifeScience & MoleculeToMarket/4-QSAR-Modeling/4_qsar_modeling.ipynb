# Notebook 4 — QSAR Modeling: Data Preparation + XGBoost + Neural Network

---

## What you will learn
QSAR = **Quantitative Structure-Activity Relationship** — predicting biological activity from
molecular structure. This notebook covers the complete QSAR pipeline in one place:

### Part A — Data Preparation
1. Fetch EGFR IC50 data from ChEMBL
2. Compute Morgan fingerprints + physicochemical descriptors
3. Deduplicate (geometric mean IC50 per compound)
4. Stratified Murcko scaffold split (train/test without scaffold leakage)

### Part B — Modeling
5. XGBoost — gradient boosted trees (fast, robust on tabular molecular data)
6. Neural network (PyTorch) — fully connected network
7. Regression: predict pIC50 (continuous potency)
8. Classification: predict active vs inactive (binary)
9. SHAP feature importance — which molecular features drive predictions?

## Why scaffold split (not random split)?
In drug discovery, molecules tested on the same target often share the same
chemical scaffold (core ring system). A random split puts similar structures
in both train and test → artificially inflated test performance.
A scaffold split ensures the test set contains **novel scaffolds**,
giving a realistic estimate of how the model generalises to new chemotypes.

**GPU recommended for neural network training, but CPU works too.**

```python
# ============================================================
# CELL 1: Install libraries
# ============================================================
print('Installing packages...')
import subprocess, sys
subprocess.run([sys.executable, '-m', 'pip', 'install', '--quiet',
                'rdkit', 'chembl-webresource-client',
                'xgboost', 'shap', 'torch',
                'pandas', 'numpy', 'matplotlib', 'scikit-learn'], check=False)
print('Done!')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

from rdkit import Chem, RDLogger
from rdkit.Chem import AllChem, Descriptors, rdMolDescriptors, DataStructs
from rdkit.Chem.Scaffolds import MurckoScaffold
RDLogger.DisableLog('rdApp.*')

from sklearn.metrics import mean_squared_error, r2_score, roc_auc_score, classification_report
from sklearn.preprocessing import StandardScaler
import xgboost as xgb
import shap
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset

print(f'XGBoost: {xgb.__version__}')
print(f'PyTorch: {torch.__version__}')
print(f'CUDA available: {torch.cuda.is_available()}')
DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'Using device: {DEVICE}')
```

---
## Part A: Data Preparation

```python
# ============================================================
# CELL 2: Fetch EGFR data from ChEMBL and compute features
# ============================================================
from chembl_webresource_client.new_client import new_client

print('Step 1: Fetching EGFR IC50 data from ChEMBL...')
print('Target: CHEMBL203 (human EGFR kinase)')

try:
    activity = new_client.activity
    raw_data = activity.filter(
        target_chembl_id='CHEMBL203',
        standard_type='IC50',
        standard_units='nM',
        standard_relation='=',
    ).only(['molecule_chembl_id', 'canonical_smiles', 'standard_value', 'standard_units'])[:500]
    df_raw = pd.DataFrame(list(raw_data))
    df_raw['standard_value'] = pd.to_numeric(df_raw['standard_value'], errors='coerce')
    df_raw = df_raw.dropna(subset=['canonical_smiles', 'standard_value'])
    df_raw = df_raw[df_raw['standard_value'] > 0]
    print(f'  Raw records fetched: {len(df_raw)}')
except Exception as e:
    print(f'ChEMBL failed: {e} — using demo subset')
    # Fallback: representative subset of EGFR inhibitors
    fallback = [
        ('CHEMBL475251', 'CCc1cc2c(cc1N)N=CN=C2Nc1cccc(Cl)c1', 15.0),
        ('CHEMBL194814', 'C=CC(=O)Nc1cccc(Nc2ncnc3ccc(OC)c(OCCCN4CCOCC4)c23)c1', 1.0),
        ('CHEMBL2110562', 'CC(=O)Nc1ccc2c(ncnc2c1O)Nc1ccc(F)cc1Cl', 2500.0),
        ('CHEMBL1',  'CC(=O)Oc1ccccc1C(=O)O', 10000.0),  # aspirin: inactive
    ] * 25  # repeat to have enough data
    df_raw = pd.DataFrame(fallback, columns=['molecule_chembl_id','canonical_smiles','standard_value'])

print('\nStep 2: Deduplicating — geometric mean IC50 per unique compound...')
# Multiple assays per compound: use geometric mean IC50
df_dedup = (df_raw.groupby('molecule_chembl_id')
            .agg(canonical_smiles=('canonical_smiles', 'first'),
                 IC50_nM=('standard_value', lambda x: np.exp(np.log(x).mean())))
            .reset_index())
print(f'  After dedup: {len(df_dedup)} unique compounds')

print('\nStep 3: Computing pIC50 (−log10 of IC50 in Molar)...')
# pIC50: higher = more potent. Easier to model than raw IC50 (compressed range).
df_dedup['pIC50'] = -np.log10(df_dedup['IC50_nM'] * 1e-9)
print(f'  pIC50 range: {df_dedup["pIC50"].min():.2f} to {df_dedup["pIC50"].max():.2f}')

print('\nStep 4: Computing Morgan fingerprints (radius=2, 1024 bits) + descriptors...')
rows = []
for _, row in df_dedup.iterrows():
    mol = Chem.MolFromSmiles(row['canonical_smiles'])
    if mol is None:
        continue
    fp  = AllChem.GetMorganFingerprintAsBitVect(mol, 2, nBits=1024)
    arr = np.zeros(1024, dtype=np.int8)
    DataStructs.ConvertToNumpyArray(fp, arr)
    rows.append({
        'chembl_id': row['molecule_chembl_id'],
        'smiles':    row['canonical_smiles'],
        'IC50_nM':   row['IC50_nM'],
        'pIC50':     row['pIC50'],
        'MW':        Descriptors.MolWt(mol),
        'LogP':      Descriptors.MolLogP(mol),
        'TPSA':      Descriptors.TPSA(mol),
        'HBD':       rdMolDescriptors.CalcNumHBD(mol),
        'HBA':       rdMolDescriptors.CalcNumHBA(mol),
        **{f'fp_{i}': int(arr[i]) for i in range(1024)},
    })
df = pd.DataFrame(rows)
print(f'  Final feature matrix: {df.shape[0]} compounds × {df.shape[1]} columns')
```

```python
# ============================================================
# CELL 3: Murcko scaffold split
# ============================================================
# Murcko scaffold = the ring systems + linker atoms of a molecule (no substituents)
# We group molecules by scaffold and ensure each scaffold appears in only ONE split.
# This prevents the model from 'memorising' scaffold patterns.

from collections import defaultdict

print('Step 5: Murcko scaffold split (train=80%, test=20%)...')

# Assign each molecule to a scaffold
scaffold_to_indices = defaultdict(list)
for i, row in df.iterrows():
    mol = Chem.MolFromSmiles(row['smiles'])
    if mol:
        scaffold = MurckoScaffold.MurckoScaffoldSmiles(mol=mol, includeChirality=False)
    else:
        scaffold = 'invalid'
    scaffold_to_indices[scaffold].append(i)

print(f'  Unique Murcko scaffolds: {len(scaffold_to_indices)}')
print(f'  Total compounds: {len(df)}')

# Stratified assignment: sort scaffolds by size, alternately assign to test
scaffolds_sorted = sorted(scaffold_to_indices.items(), key=lambda x: len(x[1]), reverse=True)
train_idx, test_idx = [], []
target_test_size = int(0.2 * len(df))

for scaffold, indices in scaffolds_sorted:
    if len(test_idx) < target_test_size:
        test_idx.extend(indices)
    else:
        train_idx.extend(indices)

df_train = df.loc[train_idx].reset_index(drop=True)
df_test  = df.loc[test_idx].reset_index(drop=True)

print(f'  Train set: {len(df_train)} compounds ({len(df_train)/len(df)*100:.1f}%)')
print(f'  Test set : {len(df_test)} compounds ({len(df_test)/len(df)*100:.1f}%)')

# ---- Prepare feature arrays ----
FP_COLS   = [c for c in df.columns if c.startswith('fp_')]
DESC_COLS = ['MW', 'LogP', 'TPSA', 'HBD', 'HBA']
FEAT_COLS = FP_COLS + DESC_COLS

X_train = df_train[FEAT_COLS].values.astype(np.float32)
y_train = df_train['pIC50'].values.astype(np.float32)
X_test  = df_test[FEAT_COLS].values.astype(np.float32)
y_test  = df_test['pIC50'].values.astype(np.float32)

# Binary labels for classification: active = pIC50 ≥ 6 (IC50 ≤ 1 µM)
y_train_cls = (y_train >= 6).astype(int)
y_test_cls  = (y_test >= 6).astype(int)

print(f'\nFeature matrix shape: {X_train.shape[1]} features ({len(FP_COLS)} fp bits + {len(DESC_COLS)} descriptors)')
print(f'Train actives: {y_train_cls.sum()}/{len(y_train_cls)} ({y_train_cls.mean()*100:.1f}%)')
print(f'Test  actives: {y_test_cls.sum()}/{len(y_test_cls)} ({y_test_cls.mean()*100:.1f}%)')
```

---
## Part B: Modeling

```python
# ============================================================
# CELL 4: XGBoost regression (predict pIC50)
# ============================================================
# XGBoost = eXtreme Gradient Boosting
# It builds an ensemble of decision trees, where each tree corrects
# the errors of the previous ones. It is fast, interpretable, and
# handles sparse data (like fingerprints) very well.

print('=== XGBoost Regression: Predict pIC50 ===')
print('Task: continuous regression — predict the actual potency value\n')

xgb_reg = xgb.XGBRegressor(
    n_estimators=300,     # number of trees
    max_depth=6,          # maximum tree depth (controls complexity)
    learning_rate=0.05,   # shrinkage rate (smaller = more robust, slower)
    subsample=0.8,        # fraction of training data used per tree
    colsample_bytree=0.8, # fraction of features used per tree
    min_child_weight=5,   # minimum samples in a leaf node (regularisation)
    random_state=42,
    n_jobs=-1,            # use all CPU cores
    verbosity=0,
)

print('Training XGBoost regressor...')
xgb_reg.fit(X_train, y_train)
print('Training complete!')

y_pred_xgb = xgb_reg.predict(X_test)
rmse_xgb   = np.sqrt(mean_squared_error(y_test, y_pred_xgb))
r2_xgb     = r2_score(y_test, y_pred_xgb)

print(f'\nXGBoost Regression Results:')
print(f'  RMSE       = {rmse_xgb:.3f} pIC50 units')
print(f'  R-squared  = {r2_xgb:.3f}')
print(f'  Interpretation: typical prediction error = {rmse_xgb:.2f} log units')

# Scatter plot: predicted vs actual
fig, ax = plt.subplots(figsize=(5.5, 5))
ax.scatter(y_test, y_pred_xgb, alpha=0.5, color='#0D9488', s=25)
lo, hi = min(y_test.min(), y_pred_xgb.min()), max(y_test.max(), y_pred_xgb.max())
ax.plot([lo, hi], [lo, hi], 'r--', lw=1.5, label='Perfect prediction')
ax.set(xlabel='Actual pIC50', ylabel='Predicted pIC50',
       title=f'XGBoost Regression\nRMSE={rmse_xgb:.3f}  R²={r2_xgb:.3f}')
ax.legend()
plt.tight_layout(); plt.show()
```

```python
# ============================================================
# CELL 5: XGBoost classification (predict active/inactive)
# ============================================================
print('=== XGBoost Classification: Active vs Inactive ===')
print('Task: binary classification — will this compound hit the target?')
print('Active threshold: pIC50 ≥ 6 (IC50 ≤ 1 µM)\n')

xgb_cls = xgb.XGBClassifier(n_estimators=300, max_depth=5, learning_rate=0.05,
                              subsample=0.8, colsample_bytree=0.8,
                              scale_pos_weight=len(y_train_cls==0)/max(y_train_cls.sum(),1),
                              random_state=42, n_jobs=-1, verbosity=0, use_label_encoder=False)
print('Training XGBoost classifier...')
xgb_cls.fit(X_train, y_train_cls)
print('Training complete!')

y_pred_cls  = xgb_cls.predict(X_test)
y_prob_cls  = xgb_cls.predict_proba(X_test)[:, 1]
auc         = roc_auc_score(y_test_cls, y_prob_cls)

print(f'\nXGBoost Classification Results:')
print(f'  ROC-AUC = {auc:.3f}  (0.5=random, 1.0=perfect)')
print('\nClassification Report:')
print(classification_report(y_test_cls, y_pred_cls, target_names=['Inactive', 'Active']))
```

```python
# ============================================================
# CELL 6: Neural network with PyTorch
# ============================================================
# Architecture: fully connected feedforward network
# Input layer: 1029 features (1024 fp bits + 5 descriptors)
# Hidden layers: 512 → 256 → 128 → 64
# Output: 1 neuron (regression) or 2 neurons (classification)
# Regularisation: Dropout (randomly zeroes neurons during training)
# Activation: ReLU

print('=== Neural Network (PyTorch): Regression ===')
print(f'Device: {DEVICE}')

class MolNet(nn.Module):
    """Fully-connected neural network for molecular property prediction."""
    def __init__(self, input_dim, hidden_dims, out_dim, dropout=0.3):
        super().__init__()
        layers = []
        in_dim = input_dim
        for h in hidden_dims:
            layers += [
                nn.Linear(in_dim, h),    # fully connected layer
                nn.BatchNorm1d(h),       # batch normalisation (stabilises training)
                nn.ReLU(),               # non-linear activation
                nn.Dropout(dropout),     # randomly zeroes p% of neurons (regularisation)
            ]
            in_dim = h
        layers.append(nn.Linear(in_dim, out_dim))
        self.net = nn.Sequential(*layers)

    def forward(self, x):
        return self.net(x)

# Scale features (important for neural networks — descriptors have very different ranges)
scaler = StandardScaler()
X_tr_sc = scaler.fit_transform(X_train)
X_te_sc = scaler.transform(X_test)

# Convert to torch tensors
X_tr_t = torch.tensor(X_tr_sc, dtype=torch.float32).to(DEVICE)
y_tr_t = torch.tensor(y_train, dtype=torch.float32).unsqueeze(1).to(DEVICE)
X_te_t = torch.tensor(X_te_sc, dtype=torch.float32).to(DEVICE)

# DataLoader for mini-batch training
train_ds = TensorDataset(X_tr_t, y_tr_t)
train_dl = DataLoader(train_ds, batch_size=64, shuffle=True)

INPUT_DIM = X_tr_sc.shape[1]
model_nn  = MolNet(INPUT_DIM, [512, 256, 128, 64], 1, dropout=0.3).to(DEVICE)
print(f'Network architecture: {INPUT_DIM} → 512 → 256 → 128 → 64 → 1')
n_params = sum(p.numel() for p in model_nn.parameters())
print(f'Total parameters: {n_params:,}')

optimizer = optim.Adam(model_nn.parameters(), lr=1e-3, weight_decay=1e-4)
criterion = nn.MSELoss()
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, patience=10)

print('\nTraining neural network (50 epochs)...')
train_losses = []
for epoch in range(50):
    model_nn.train()
    epoch_loss = 0
    for xb, yb in train_dl:
        optimizer.zero_grad()
        pred = model_nn(xb)
        loss = criterion(pred, yb)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()
    epoch_loss /= len(train_dl)
    train_losses.append(epoch_loss)
    scheduler.step(epoch_loss)
    if (epoch + 1) % 10 == 0:
        print(f'  Epoch {epoch+1:3d}/50 | Loss={epoch_loss:.4f}')

model_nn.eval()
with torch.no_grad():
    y_pred_nn = model_nn(X_te_t).cpu().numpy().flatten()

rmse_nn = np.sqrt(mean_squared_error(y_test, y_pred_nn))
r2_nn   = r2_score(y_test, y_pred_nn)
print(f'\nNeural Network Regression Results:')
print(f'  RMSE      = {rmse_nn:.3f} pIC50 units')
print(f'  R-squared = {r2_nn:.3f}')

print('\n=== Model Comparison ===')
print(f'  XGBoost:        RMSE={rmse_xgb:.3f}  R²={r2_xgb:.3f}')
print(f'  Neural Network: RMSE={rmse_nn:.3f}  R²={r2_nn:.3f}')
```

```python
# ============================================================
# CELL 7: SHAP feature importance (XGBoost)
# ============================================================
# SHAP (SHapley Additive exPlanations) explains WHY the model made each prediction.
# SHAP values tell you: for this compound, which molecular features pushed the
# predicted pIC50 up or down?

print('=== SHAP Feature Importance Analysis ===')
print('SHAP values explain model predictions in terms of input features.')
print('Positive SHAP = feature pushes prediction HIGHER (more potent)')
print('Negative SHAP = feature pushes prediction LOWER (less potent)\n')

print('Computing SHAP values (may take 1–2 minutes)...')
explainer   = shap.TreeExplainer(xgb_reg)
shap_values = explainer.shap_values(X_test[:50])   # use 50 test examples for speed

print(f'SHAP matrix shape: {shap_values.shape}  (samples × features)')
print('Each cell = SHAP value = contribution of that feature to that prediction')

# Top features by mean |SHAP|
mean_abs_shap = np.abs(shap_values).mean(axis=0)
top_idx       = np.argsort(mean_abs_shap)[::-1][:20]
feat_names    = FEAT_COLS

print('\nTop 10 most important features (by mean |SHAP|):')
for rank, idx in enumerate(top_idx[:10]):
    fname = feat_names[idx]
    importance = mean_abs_shap[idx]
    ftype = 'fingerprint' if fname.startswith('fp_') else 'descriptor'
    print(f'  #{rank+1:2d}  {fname:<15}  mean|SHAP|={importance:.4f}  ({ftype})')

# SHAP summary plot
print('\nGenerating SHAP summary plot...')
plt.figure(figsize=(8, 5))
shap.summary_plot(shap_values, X_test[:50],
                  feature_names=feat_names,
                  max_display=15,
                  plot_type='bar', show=False)
plt.title('SHAP Feature Importance: XGBoost pIC50 Prediction', fontsize=12)
plt.tight_layout(); plt.show()

print('\n=== Notebook 4 Complete ===')

print('We will use the trained model to screen large compound libraries.')
```
