## What you will learn
ADMET = **Absorption, Distribution, Metabolism, Excretion, Toxicity**

A compound can be highly potent at a target but still fail as a drug if it:
- Is poorly absorbed after oral dosing (bad Absorption)
- Cannot reach the target tissue (bad Distribution)
- Is metabolised too quickly (bad Metabolism)
- Cannot be cleared from the body (bad Excretion)
- Is toxic to organs (bad Toxicity)

**90% of compounds that enter clinical trials FAIL**, and many of these failures
are due to ADMET problems that could have been predicted computationally!

```python
# ============================================================
# CELL 2: ADMET properties overview
# ============================================================
print('=== ADMET Properties: What They Are and Why They Matter ===')
print()
print('A = ABSORPTION')
print('  - Oral bioavailability: how much of the drug reaches systemic circulation')
print('  - Lipinski Ro5, Veber rules, Caco-2 permeability, P-glycoprotein efflux')
print()
print('D = DISTRIBUTION')
print('  - Volume of distribution: how widely the drug spreads through the body')
print('  - Blood-brain barrier (BBB) penetration (important for CNS drugs)')
print('  - Plasma protein binding (affects free drug concentration)')
print()
print('M = METABOLISM')
print('  - CYP450 enzyme inhibition/induction (drug-drug interaction risk)')
print('  - Half-life: how long the drug stays active in the body')
print('  - Reactive metabolite formation (toxicity risk)')
print()
print('E = EXCRETION')
print('  - Renal clearance, hepatic clearance')
print('  - Total clearance determines dosing frequency')
print()
print('T = TOXICITY')
print('  - hERG inhibition → cardiac arrhythmia risk (QT prolongation)')
print('  - Ames mutagenicity, hepatotoxicity, skin sensitisation')
print()
print('NOTE: These properties are strongly linked to molecular structure.')
print('Modern ML models can predict them from SMILES with reasonable accuracy.')
```

```python
# ============================================================
# CELL 3: Rule-based ADMET filters
# ============================================================
# Several rule-based filters are commonly applied before ML prediction:
#   Lipinski Ro5    — oral bioavailability filter
#   Veber rules     — additional oral bioavailability filter
#   PAINS filter    — Pan-Assay INterference compoundS (false positives)

DRUG_SMILES = {
    'Erlotinib':     'COc1cc2ncnc(Nc3cccc(C#C)c3)c2cc1OCCO',
    'Osimertinib':   'COc1cc2ncnc(Nc3cccc(NC(=O)/C=C/CN(C)C)c3)c2cc1OCCCN1CCOCC1',
    'Aspirin':       'CC(=O)Oc1ccccc1C(=O)O',
    'Doxorubicin':   'COc1cccc2c1C(=O)c1c(O)c3c(c(O)c1C2=O)C[C@@](O)(C(=O)CO)C[C@H]3O[C@H]1C[C@H](N)[C@H](O)[C@H](C)O1',
    'PAINS_example': 'O=C1N(CC2=CC=CC=C2)C2=CC=CC=C2C1=O',  # thalidomide-like (PAINS alert)
}

# Set up PAINS filter using RDKit FilterCatalog
pains_params = FilterCatalogParams()
pains_params.AddCatalog(FilterCatalogParams.FilterCatalogs.PAINS)
pains_catalog = FilterCatalog.FilterCatalog(pains_params)

print('=== Rule-Based ADMET Filters ===')
print(f'{"Compound":<20} {"MW":>6} {"LogP":>6} {"TPSA":>6} {"RotB":>6} {"Ro5":>6} {"Veber":>6} {"PAINS":>8}')
print('-' * 80)

results = []
for name, smiles in DRUG_SMILES.items():
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        print(f'{name}: invalid SMILES')
        continue

    mw   = Descriptors.MolWt(mol)
    logp = Descriptors.MolLogP(mol)
    tpsa = Descriptors.TPSA(mol)
    rotb = rdMolDescriptors.CalcNumRotatableBonds(mol)
    hbd  = rdMolDescriptors.CalcNumHBD(mol)
    hba  = rdMolDescriptors.CalcNumHBA(mol)

    # Lipinski Ro5: ≤1 violation
    ro5_viol = sum([mw > 500, logp > 5, hbd > 5, hba > 10])
    ro5_pass = 'PASS' if ro5_viol <= 1 else 'FAIL'

    # Veber rules: rotatable bonds ≤10 AND TPSA ≤140
    veber_pass = 'PASS' if (rotb <= 10 and tpsa <= 140) else 'FAIL'

    # PAINS filter
    pains_hit = pains_catalog.HasMatch(mol)
    pains_str = 'ALERT' if pains_hit else 'clean'

    print(f'{name:<20} {mw:6.0f} {logp:6.2f} {tpsa:6.1f} {rotb:6d} {ro5_pass:>6} {veber_pass:>6} {pains_str:>8}')
    results.append({'Compound': name, 'MW': mw, 'LogP': logp, 'TPSA': tpsa,
                    'RotB': rotb, 'Ro5': ro5_pass, 'Veber': veber_pass, 'PAINS': pains_str})

print('\nNote: PAINS compounds can cause false positives in assays by binding non-specifically.')
print('They should be flagged, not necessarily excluded, but validated carefully.')
```

```python
# ============================================================
# CELL 4: ML-based ADMET prediction (simulated)
# ============================================================
# Real tools: ADMET-AI, pkCSM, SwissADME, DeepPurpose
# Here we simulate predictions to show the workflow without API calls.
# For production use, substitute with actual ADMET prediction tools.

print('=== ML-Based ADMET Property Prediction ===')
print('In practice: use pkCSM, SwissADME, ADMET-AI, or DeepPurpose APIs')
print('Here we demonstrate the workflow with rule-based approximations\n')

def predict_admet_simple(smiles):
    """
    Simple rule-based ADMET estimates.
    Returns dict of ADMET properties (0-1 scale where possible).
    Real tools use trained ML models on experimental datasets.
    """
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        return None

    mw   = Descriptors.MolWt(mol)
    logp = Descriptors.MolLogP(mol)
    tpsa = Descriptors.TPSA(mol)
    hbd  = rdMolDescriptors.CalcNumHBD(mol)
    hba  = rdMolDescriptors.CalcNumHBA(mol)
    rotb = rdMolDescriptors.CalcNumRotatableBonds(mol)
    arom = rdMolDescriptors.CalcNumAromaticRings(mol)

    # Oral absorption score (0=poor, 1=excellent) — Lipinski + Veber proxy
    ro5_viol = sum([mw > 500, logp > 5, hbd > 5, hba > 10])
    absorption = max(0, 1.0 - ro5_viol * 0.4 - (tpsa - 90) / 200 if tpsa > 90 else 1.0)

    # BBB penetration proxy: logP 1-3 and MW < 400 and TPSA < 90 → good BBB
    bbb = 1.0 if (1 < logp < 3 and mw < 400 and tpsa < 90) else 0.3

    # hERG toxicity risk: aromatic + logP > 3 → higher risk
    herg_risk = min(1.0, (arom * 0.15 + max(0, logp - 2) * 0.15))

    # Metabolic stability proxy: heavy atoms + logP
    n_heavy = mol.GetNumHeavyAtoms()
    stab = max(0, 1.0 - n_heavy / 60 - max(0, logp - 3) / 10)

    # Solubility (ESOL approximation): lower logP and MW → better solubility
    solubility = max(0, 1.0 - logp / 5 - mw / 800)

    return {
        'Absorption':  round(absorption, 2),
        'BBB_penetration': round(bbb, 2),
        'Metabolic_stability': round(stab, 2),
        'Solubility':  round(solubility, 2),
        'hERG_risk':   round(herg_risk, 2),
    }

print(f'{"Compound":<20}  Absorption  BBB  Metabolism  Solubility  hERG-risk')
print('-' * 80)
admet_results = {}
for name, smiles in DRUG_SMILES.items():
    props = predict_admet_simple(smiles)
    if props:
        admet_results[name] = props
        print(f'{name:<20}  {props["Absorption"]:10.2f}  {props["BBB_penetration"]:5.2f}  '
              f'{props["Metabolic_stability"]:10.2f}  {props["Solubility"]:10.2f}  {props["hERG_risk"]:9.2f}')
```

```python
# ============================================================
# CELL 5: Radar (spider) chart visualisation of ADMET profiles
# ============================================================
# Radar charts show multiple ADMET properties simultaneously,
# making it easy to compare drugs across all dimensions at once.

print('=== ADMET Radar Chart Visualisation ===')
print('Radar charts show all 5 ADMET dimensions simultaneously.')
print('NOTE: for hERG_risk we invert (1-risk) so that larger = better in all dimensions\n')

RADAR_PROPS  = ['Absorption', 'BBB_penetration', 'Metabolic_stability', 'Solubility', 'hERG_safety']
RADAR_LABELS = ['Absorption', 'BBB Penetration', 'Metabolic\nStability', 'Solubility', 'hERG Safety\n(1−risk)']
COLORS = ['#0D9488', '#1B2A4A', '#E57373', '#F59E0B', '#6B7280']

fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))

n_props  = len(RADAR_PROPS)
angles   = [n / float(n_props) * 2 * np.pi for n in range(n_props)]
angles  += angles[:1]  # close the polygon

ax.set_xticks(angles[:-1])
ax.set_xticklabels(RADAR_LABELS, fontsize=10)
ax.set_ylim(0, 1.05)
ax.set_yticks([0.25, 0.5, 0.75, 1.0])
ax.set_yticklabels(['0.25', '0.50', '0.75', '1.0'], fontsize=7, color='gray')

for (name, props), color in zip(admet_results.items(), COLORS):
    values = [
        props['Absorption'],
        props['BBB_penetration'],
        props['Metabolic_stability'],
        props['Solubility'],
        1.0 - props['hERG_risk'],  # invert: 1−risk so bigger = safer
    ]
    values += values[:1]
    ax.plot(angles, values, 'o-', color=color, lw=2.0, label=name)
    ax.fill(angles, values, alpha=0.08, color=color)

ax.set_title('ADMET Radar Profiles — Bigger Area = Better Drug-Like Properties',
             fontsize=12, fontweight='bold', pad=25)
ax.legend(loc='upper right', bbox_to_anchor=(1.35, 1.1), fontsize=9)
plt.tight_layout()
plt.show()

print('\n=== Notebook 7 Complete ===')

print('We will design new drug-like molecules using generative AI.')
```
