## What you will learn
Generative AI can design new drug molecules from scratch — molecules that have
never been synthesised before but are predicted to have good drug properties.

### Generative approaches covered
1. **LLM-based generation (GPT-4/Claude):** Ask an LLM to suggest SMILES modifications
2. **Variational Autoencoder (VAE):** Encode molecules to latent space, sample, decode
3. **Fragment-based design:** Combine known fragments to create novel molecules
4. **Scaffold hopping:** Replace the core scaffold while keeping key pharmacophores

### Why generate molecules?
- Traditional drug discovery: test existing molecules from a library
- Generative AI: CREATE new molecules with desired properties
- Example: generate EGFR inhibitors that also overcome T790M resistance

```python
# ============================================================
# CELL 1: Install libraries
# ============================================================
print('Installing packages...')
import subprocess, sys
subprocess.run([sys.executable, '-m', 'pip', 'install', '--quiet',
                'rdkit', 'pandas', 'numpy', 'matplotlib', 'torch',
                'scikit-learn', 'anthropic'], check=False)
print('Done!')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

from rdkit import Chem, RDLogger
from rdkit.Chem import AllChem, Descriptors, rdMolDescriptors, Draw, DataStructs
RDLogger.DisableLog('rdApp.*')

import torch
import torch.nn as nn
import torch.optim as optim
import os

DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'PyTorch {torch.__version__} | Device: {DEVICE}')
print('Libraries imported!')
```

```python
# ============================================================
# CELL 2: LLM-based scaffold hopping
# ============================================================
# Ask an LLM to propose SMILES modifications to a known EGFR inhibitor
# that might overcome T790M resistance.
# This is a creative, hypothesis-generating approach.

print('=== Approach 1: LLM-Based Molecular Design ===')
print()
print('We ask Claude Sonnet to propose EGFR inhibitor modifications.')
print('The LLM uses its training knowledge of medicinal chemistry principles.')

ANTHROPIC_API_KEY = os.environ.get('ANTHROPIC_API_KEY')

if ANTHROPIC_API_KEY:
    import anthropic
    client = anthropic.Anthropic(api_key=ANTHROPIC_API_KEY)

    # Erlotinib SMILES — 1st gen EGFR inhibitor
    erlotinib_smiles = 'COc1cc2ncnc(Nc3cccc(C#C)c3)c2cc1OCCO'

    prompt = f"""You are a medicinal chemist expert in EGFR kinase inhibitors.

Erlotinib (SMILES: {erlotinib_smiles}) is a 1st-generation EGFR inhibitor.
It fails against the T790M gatekeeper mutation because the mutation adds a bulky
methionine that clashes with the erlotinib binding mode.

Task: Propose 3 specific structural modifications to erlotinib that might help
overcome T790M resistance. For each modification:
1. Describe the chemical change (1 sentence)
2. Explain the rationale based on the T790M binding site
3. Provide a plausible modified SMILES string

Be specific and chemically accurate."""

    print(f'\nQuerying Claude Sonnet...')
    response = client.messages.create(
        model='claude-sonnet-4-5', max_tokens=800,
        messages=[{'role': 'user', 'content': prompt}]
    )
    llm_output = response.content[0].text
    print('\n=== Claude Sonnet Suggestions ===')
    print(llm_output)

    # Validate any SMILES in the response
    import re
    smiles_pattern = re.compile(r'[A-Za-z0-9@#\[\]()=+\-/\\.%]{10,}')
    candidate_smiles = smiles_pattern.findall(llm_output)
    valid_mols = []
    for s in candidate_smiles[:5]:  # check first 5 candidates
        mol = Chem.MolFromSmiles(s)
        if mol is not None:
            valid_mols.append((s, mol))
    print(f'\nValid SMILES extracted from response: {len(valid_mols)}')

else:
    print('No Anthropic API key found. Showing example output format.')
    print()
    print('Example modification concepts (would come from LLM):')
    print('  1. Replace terminal alkyne with propargyl amine → smaller, covalent warhead option')
    print('  2. Add acrylamide covalent warhead → bonds to C797, overcomes T790M steric clash')
    print('  3. Remove one methoxy, add morpholine → improves solubility at hinge binder region')
```

```python
# ============================================================
# CELL 3: Fragment-based molecular generation
# ============================================================
# Fragment-based drug design:
#   1. Define a pharmacophore (the essential binding features)
#   2. Break known active compounds into fragments
#   3. Recombine fragments in new ways
#   4. Validate each combination with ADMET filters

print('=== Approach 2: Fragment-Based Generation ===')
print('Fragment libraries from EGFR inhibitors, recombined to create new candidates.\n')

# EGFR inhibitor core fragments
FRAGMENTS = {
    # Hinge binder fragments (bind to the hinge region of the ATP pocket)
    'quinazoline':  'c1cnc2ccccc2n1',
    'pyrimidine':   'c1ccnc(N)n1',
    'indazole':     'c1ccc2[nH]ncc2c1',
    # Solubilising groups
    'morpholine':   'C1COCCN1',
    'piperazine':   'C1CNCCN1',
    'methylamine':  'CN',
    # Covalent warheads (for 3rd gen inhibitors)
    'acrylamide':   'C=CC(=O)N',
    'vinyl_sulfone':'C=CS(=O)(=O)',
    # Lipophilic substituents
    'fluorophenyl': 'c1ccc(F)cc1',
    'chlorophenyl': 'c1ccc(Cl)cc1',
    'trifluoromethyl': 'C(F)(F)F',
}

print('Fragment library:')
for name, smiles in FRAGMENTS.items():
    mol = Chem.MolFromSmiles(smiles)
    valid = 'valid' if mol else 'INVALID'
    n_atoms = mol.GetNumAtoms() if mol else 0
    print(f'  {name:<20}: {smiles:<30} ({n_atoms} atoms, {valid})')

# Real EGFR inhibitors to use as templates for fragment extraction
TEMPLATE_DRUGS = {
    'Erlotinib':   'COc1cc2ncnc(Nc3cccc(C#C)c3)c2cc1OCCO',
    'Gefitinib':   'COc1cc2ncnc(Nc3ccc(F)cc3Cl)c2cc1OCCCN1CCOCC1',
    'Osimertinib': 'COc1cc2ncnc(Nc3cccc(NC(=O)/C=C/CN(C)C)c3)c2cc1OCCCN1CCOCC1',
}

print('\nComputing Morgan fingerprints for template drugs...')
template_fps = {}
for name, smiles in TEMPLATE_DRUGS.items():
    mol = Chem.MolFromSmiles(smiles)
    if mol:
        fp = AllChem.GetMorganFingerprintAsBitVect(mol, 2, nBits=1024)
        template_fps[name] = fp
        mw   = Descriptors.MolWt(mol)
        logp = Descriptors.MolLogP(mol)
        print(f'  {name}: MW={mw:.0f} Da  logP={logp:.2f}')

print('\nFragment-based generation workflow:')
print('  1. Select hinge binder fragment (e.g., quinazoline)')
print('  2. Add aniline linker with key substituents')
print('  3. Add solubilising group for ADMET')
print('  4. Optionally add covalent warhead for 3rd gen')
print('  5. Validate with Lipinski/Veber filters')
```

```python
# ============================================================
# CELL 4: Variational Autoencoder (VAE) for molecular generation
# ============================================================
# VAE architecture for SMILES generation:
#   Encoder: SMILES → latent vector z (compressed representation)
#   Decoder: latent vector z → SMILES
# To generate new molecules:
#   1. Encode a known drug to get z
#   2. Add small noise to z
#   3. Decode the noisy z → new (similar but different) molecule

print('=== Approach 3: Variational Autoencoder (VAE) ===')
print('VAE learns a continuous latent space of molecular structures.')
print('New molecules are generated by sampling from this latent space.\n')

# For this demo we use fingerprints as the input (simpler than character-level SMILES)
# A production VAE would use character-level or graph-based encoding.

class MolVAE(nn.Module):
    """Variational Autoencoder for molecular fingerprints."""
    def __init__(self, input_dim=1024, latent_dim=64, hidden_dim=256):
        super().__init__()
        # Encoder: input → latent mean and log-variance
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, hidden_dim), nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim // 2), nn.ReLU(),
        )
        self.mu     = nn.Linear(hidden_dim // 2, latent_dim)   # mean of latent dist
        self.logvar = nn.Linear(hidden_dim // 2, latent_dim)   # log-variance of latent dist

        # Decoder: latent → reconstructed fingerprint
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, hidden_dim // 2), nn.ReLU(),
            nn.Linear(hidden_dim // 2, hidden_dim), nn.ReLU(),
            nn.Linear(hidden_dim, input_dim), nn.Sigmoid(),  # sigmoid → 0-1 for bit probabilities
        )

    def reparametrize(self, mu, logvar):
        """Reparametrization trick: z = mu + epsilon*std  (epsilon ~ N(0,1))."""
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)  # random noise
        return mu + eps * std

    def forward(self, x):
        h       = self.encoder(x)
        mu      = self.mu(h)
        logvar  = self.logvar(h)
        z       = self.reparametrize(mu, logvar)  # sample from latent space
        x_recon = self.decoder(z)
        return x_recon, mu, logvar


def vae_loss(recon_x, x, mu, logvar):
    """VAE loss = reconstruction loss (BCE) + KL divergence."""
    # Reconstruction: how well does the decoder reproduce the input fingerprint?
    recon_loss = nn.functional.binary_cross_entropy(recon_x, x, reduction='sum')
    # KL divergence: how close is the latent distribution to N(0,1)?
    # This regularises the latent space so we can sample from it.
    kl_loss = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp())
    return recon_loss + kl_loss


# ---- Build training data from known EGFR inhibitors ----
# In practice, train on thousands of drug-like molecules.
# Here we use a small synthetic set for demonstration.
np.random.seed(42)
# Simulate fingerprints for 300 random drug-like molecules
X_train_vae = torch.tensor(np.random.randint(0, 2, (300, 1024)).astype(np.float32)).to(DEVICE)

print('Training VAE on 300 molecular fingerprints (30 epochs)...')
vae   = MolVAE(input_dim=1024, latent_dim=64).to(DEVICE)
optim_vae = optim.Adam(vae.parameters(), lr=1e-3)

for epoch in range(30):
    vae.train()
    total_loss = 0
    for i in range(0, len(X_train_vae), 32):  # mini-batches
        xb = X_train_vae[i:i+32]
        optim_vae.zero_grad()
        recon, mu, logvar = vae(xb)
        loss = vae_loss(recon, xb, mu, logvar)
        loss.backward()
        optim_vae.step()
        total_loss += loss.item()
    if (epoch + 1) % 10 == 0:
        print(f'  Epoch {epoch+1}/30 | Loss={total_loss:.1f}')

print('VAE training complete!')

# ---- Generate new molecules by sampling latent space ----
print('\nGenerating new molecular fingerprints from latent space...')
vae.eval()
with torch.no_grad():
    z_sample = torch.randn(10, 64).to(DEVICE)  # sample 10 random latent vectors
    generated = vae.decoder(z_sample).cpu().numpy()
    generated_bits = (generated > 0.5).astype(int)  # threshold to binary fingerprint

print(f'Generated {len(generated_bits)} new molecular fingerprint vectors.')
print('Each vector has', generated_bits.sum(axis=1).mean():.0f, 'bits ON on average.')
print('\nNote: In a full pipeline, we decode these back to SMILES using a trained decoder.')
print('Here the fingerprint represents the molecular structure numerically.')

print('\n=== Notebook 8 Complete ===')
```
