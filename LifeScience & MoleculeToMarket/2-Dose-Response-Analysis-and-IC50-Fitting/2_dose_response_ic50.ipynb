## Background: Why IC50 matters
IC50 (half-maximal inhibitory concentration) is the most important number from a
drug screening experiment. It tells you how potent a compound is at inhibiting a
biological target. Lower IC50 = more potent compound.

In a standard cell viability assay:
- **100% viability** = no drug effect (vehicle control)
- **0% viability** = complete inhibition / cell death
- **IC50** = drug concentration causing 50% reduction in viability

## Cell 4: The 4-Parameter Logistic (4PL) Model

The **4PL model** is the gold standard for dose-response analysis in drug discovery.
It is a sigmoid (S-shaped) function with four parameters:

$$
\text{Response} = \text{Bottom} + \frac{\text{Top} - \text{Bottom}}{1 + \left(\dfrac{\text{IC}_{50}}{\text{Concentration}}\right)^{\text{Hill}}}
$$

| Parameter | Biological meaning | Typical value |
|-----------|-------------------|---------------|
| **Bottom** | Maximal inhibition (high drug concentration) | ~0% viability |
| **Top** | Baseline (no drug = vehicle control) | ~100% viability |
| **IC50** | Concentration at 50% effect | depends on compound |
| **Hill** | Steepness / cooperativity of the transition | 0.5 to 3 in practice |

**Why 4PL and not simpler models?**
- Simple linear regression cannot capture the sigmoid shape
- 2PL (fix Bottom=0, Top=100) fails when inhibition is incomplete
- 4PL handles asymmetric curves and partial inhibition
- It is interpretable: each parameter maps directly to biology

```python
# ============================================================
# CELL 4: Define and visualise the 4PL model
# ============================================================

def four_pl(concentration, bottom, top, ic50, hill):
    """
    4-Parameter Logistic (4PL) dose-response model.

    Parameters
    ----------
    concentration : array of drug concentrations (µM)
    bottom        : response at saturating drug concentration (fully inhibited)
    top           : response at zero drug (baseline / vehicle control)
    ic50          : half-maximal inhibitory concentration (µM)
    hill          : Hill slope (steepness of the sigmoid)

    Returns
    -------
    response : predicted response values (%)

    Note: In a viability assay the curve goes DOWN (high conc → low viability),
    so bottom < top. This is handled automatically by the formula.
    """
    return bottom + (top - bottom) / (1.0 + (ic50 / concentration) ** hill)


# ---- Demonstrate how each parameter shapes the curve ----
conc_range = np.logspace(-3, 2, 300)   # 0.001 to 100 µM

print('Generating parameter effect plots...')
print('These plots help build intuition for what each 4PL parameter controls.')

fig, axes = plt.subplots(1, 3, figsize=(14, 4.5))

# Panel 1: Effect of IC50
ax = axes[0]
for ic50_val, color, lw in [(0.1, '#E57373', 2.5), (1.0, '#1B2A4A', 2.5), (10.0, '#0D9488', 2.5)]:
    resp = four_pl(conc_range, bottom=0, top=100, ic50=ic50_val, hill=1.5)
    ax.semilogx(conc_range, resp, color=color, lw=lw, label=f'IC50 = {ic50_val} µM')
ax.axhline(50, color='gray', lw=0.8, ls='--', alpha=0.5)
ax.set(xlabel='Concentration (µM)', ylabel='Viability (%)',
       title='Effect of IC50', ylim=(-5, 110))
ax.legend(fontsize=9)
print('  Panel 1: IC50 shifts curve LEFT (more potent) or RIGHT (less potent)')

# Panel 2: Effect of Hill slope
ax = axes[1]
for hill_val, color in [(0.5, '#E57373'), (1.5, '#1B2A4A'), (3.0, '#0D9488')]:
    resp = four_pl(conc_range, bottom=0, top=100, ic50=1.0, hill=hill_val)
    ax.semilogx(conc_range, resp, color=color, lw=2.5, label=f'Hill = {hill_val}')
ax.axhline(50, color='gray', lw=0.8, ls='--', alpha=0.5)
ax.set(xlabel='Concentration (µM)', title='Effect of Hill Slope', ylim=(-5, 110))
ax.legend(fontsize=9)
print('  Panel 2: Hill slope controls steepness (higher = sharper transition)')

# Panel 3: Effect of Bottom plateau (incomplete inhibition)
ax = axes[2]
for floor, color in [(0, '#0D9488'), (30, '#1B2A4A'), (60, '#E57373')]:
    resp = four_pl(conc_range, bottom=100, top=floor, ic50=1.0, hill=1.5)
    ax.semilogx(conc_range, resp, color=color, lw=2.5, label=f'Min viability = {floor}%')
ax.axhline(50, color='gray', lw=0.8, ls='--', alpha=0.5)
ax.set(xlabel='Concentration (µM)', title='Effect of Bottom Plateau', ylim=(-5, 110))
ax.legend(fontsize=9)
print('  Panel 3: Bottom plateau = floor viability at saturation (0% = full kill)')

fig.suptitle('4PL Model: How Each Parameter Shapes the Dose-Response Curve',
             fontsize=12, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
print('\nKey takeaway: IC50 is the most important parameter for ranking compound potency.')
```

## Cell 6: 4PL curve fitting function

**`scipy.optimize.curve_fit`** uses nonlinear least squares (Levenberg-Marquardt algorithm)
to find the 4 parameters that minimise the sum of squared residuals.

**Critical: initial parameter guesses (p0)**
Nonlinear optimisation can get stuck in local minima. Good starting guesses prevent this:
- **Bottom**: use the 5th percentile of observed responses (robust minimum)
- **Top**: use the 95th percentile (robust maximum)
- **IC50**: use geometric mean of concentration range (midpoint on log scale)
- **Hill**: start at 1.0 (neutral slope, no prior assumption)

**95% confidence intervals on IC50**
The covariance matrix returned by `curve_fit` gives standard errors on each parameter.
We use the t-distribution critical value with `n − 4` degrees of freedom.

```python
# ============================================================
# CELL 6: 4PL fitting function with IC50 confidence intervals
# ============================================================

def make_p0(conc, resp):
    """
    Data-driven initial parameter guesses for 4PL fitting.
    Good starting values are essential for nonlinear optimisation.

    Returns [bottom, top, ic50, hill]
    """
    bottom = float(np.percentile(resp, 5))               # robust minimum
    top    = float(np.percentile(resp, 95))              # robust maximum
    ic50   = float(np.exp(np.mean(np.log(conc))))        # geometric mean of conc range
    hill   = 1.0                                          # neutral starting slope
    return [bottom, top, ic50, hill]


def fit_dose_response(conc, resp):
    """
    Fit a 4PL model to dose-response data.

    Parameters
    ----------
    conc : array of concentrations (µM)
    resp : array of response values (%)

    Returns
    -------
    dict with:
        success   : bool — did fitting converge?
        params    : [bottom, top, ic50, hill]
        ic50      : fitted IC50 (µM)
        ic50_ci   : (lower, upper) 95% confidence interval on IC50
        r_squared : R² (coefficient of determination, 0–1; 1 = perfect fit)
        rmse      : root mean squared error (%)
        message   : status string
    """
    result = dict(success=False, params=None, pcov=None,
                  ic50=None, ic50_ci=(None, None),
                  r_squared=None, rmse=None, message='')

    conc = np.asarray(conc, dtype=float)
    resp = np.asarray(resp, dtype=float)

    p0 = make_p0(conc, resp)
    print(f'  Initial guess: bottom={p0[0]:.1f}  top={p0[1]:.1f}  ic50={p0[2]:.3f}  hill={p0[3]:.1f}')

    # Parameter bounds: prevent physically impossible values
    # bottom: allow slightly negative (measurement noise can go below 0)
    # top: allow slightly above 100% (noise can push above baseline)
    # ic50: must be positive; allow outside tested range
    # hill: typical range 0.1 to 10
    bounds = ([-20, -20, conc.min() * 0.01, 0.1],
              [150,  150, conc.max() * 100,  10.0])

    try:
        with warnings.catch_warnings():
            warnings.simplefilter('ignore')
            popt, pcov = curve_fit(
                four_pl, conc, resp,
                p0=p0, bounds=bounds,
                maxfev=10000,          # maximum function evaluations
            )
    except Exception as e:
        result['message'] = f'Fitting failed: {e}'
        print(f'  FAILED: {e}')
        return result

    bottom, top, ic50, hill = popt

    # ---- Goodness of fit: R² ----
    # R² = 1 - (sum of squared residuals) / (total sum of squares)
    # R² = 1.0 means perfect fit; R² near 0 means no better than predicting the mean
    resp_pred = four_pl(conc, *popt)
    ss_res    = np.sum((resp - resp_pred) ** 2)
    ss_tot    = np.sum((resp - resp.mean()) ** 2)
    r2        = 1.0 - ss_res / ss_tot if ss_tot > 0 else 0.0

    # ---- RMSE: typical error per data point ----
    rmse = np.sqrt(ss_res / len(resp))

    # ---- 95% Confidence interval on IC50 ----
    # pcov is the covariance matrix; diagonal = variance of each parameter
    # SE = sqrt(variance); CI = SE × t-critical (two-tailed, α=0.05)
    perr     = np.sqrt(np.diag(pcov))          # standard errors
    df_resid = max(len(resp) - 4, 1)           # degrees of freedom = n - num_params
    t_crit   = t_dist.ppf(0.975, df=df_resid)  # 95% CI critical value
    ic50_lo  = ic50 - t_crit * perr[2]         # index 2 = ic50 in param vector
    ic50_hi  = ic50 + t_crit * perr[2]

    result.update({
        'success':   True,
        'params':    popt,
        'pcov':      pcov,
        'ic50':      ic50,
        'ic50_ci':   (max(ic50_lo, 0.0), ic50_hi),  # IC50 cannot be negative
        'r_squared': r2,
        'rmse':      rmse,
        'message':   'Converged',
    })
    return result


print('Fitting functions defined.')
print()
print('=== Fitting all 5 compounds ===')

fit_results = {}
for name, df in data.items():
    conc = df['concentration_uM'].values
    resp = df['response_pct'].values
    print(f'\n{name}:')
    fit = fit_dose_response(conc, resp)
    fit_results[name] = fit

    if fit['success']:
        p = fit['params']
        print(f'  Converged: Bottom={p[0]:.1f}%  Top={p[1]:.1f}%  IC50={fit["ic50"]:.3f}µM  Hill={p[3]:.2f}')
        print(f'  IC50 95% CI: ({fit["ic50_ci"][0]:.3f}, {fit["ic50_ci"][1]:.3f}) µM')
        print(f'  R² = {fit["r_squared"]:.3f}  |  RMSE = {fit["rmse"]:.2f}%')
    else:
        print(f'  {fit["message"]}')
```

## Cell 7: Visualise fitted curves

For each compound we plot:
- Raw data points (circles)
- Fitted 4PL curve (solid line)
- IC50 marker (dashed crosshair)
- R² and RMSE in the corner

Look carefully at Compound C (hook effect) and D (incomplete) —
the IC50 numbers appear, but the visual shows they should not be trusted!

```python
# ============================================================
# CELL 7: Plot fitted curves for all five compounds
# ============================================================

def plot_one_curve(ax, name, df, fit, color):
    """
    Plot raw data + fitted 4PL curve on a given axes object.
    Also shows IC50 marker and quality metrics.
    """
    conc = df['concentration_uM'].values
    resp = df['response_pct'].values

    # Raw observations
    ax.semilogx(conc, resp, 'o', color=color, ms=7, alpha=0.85, zorder=3, label='Data')

    if fit['success']:
        # Smooth fitted curve (300 points for clean line)
        conc_fine = np.logspace(np.log10(conc.min()) - 0.3,
                                np.log10(conc.max()) + 0.3, 300)
        resp_fit  = four_pl(conc_fine, *fit['params'])
        ax.semilogx(conc_fine, resp_fit, '-', color=color, lw=2.5, alpha=0.9, label='4PL fit')

        # IC50 crosshair
        ic50          = fit['ic50']
        resp_at_ic50  = four_pl(np.array([ic50]), *fit['params'])[0]
        ax.axvline(ic50, color='gray', lw=1.0, ls='--', alpha=0.5)
        ax.axhline(resp_at_ic50, color='gray', lw=1.0, ls='--', alpha=0.5)
        ax.annotate(
            f'IC50\n{ic50:.2f}µM',
            xy=(ic50, resp_at_ic50),
            xytext=(ic50 * 4, resp_at_ic50 + 15),
            fontsize=8, color='gray',
            arrowprops=dict(arrowstyle='->', color='gray', lw=0.8),
        )

        # Quality metrics in corner
        ax.text(0.03, 0.08,
                f'R²={fit["r_squared"]:.2f}  RMSE={fit["rmse"]:.1f}%',
                transform=ax.transAxes, fontsize=8, color='#555')
    else:
        ax.text(0.5, 0.5, 'Fit\nFailed', transform=ax.transAxes,
                ha='center', color='red', fontsize=12, fontweight='bold')

    short    = name.split('(')[0].strip()
    scenario = name[name.find('(')+1:name.find(')')]
    ax.set_title(f'{short}\n({scenario})', fontsize=10, fontweight='bold')
    ax.set(xlabel='Conc (µM)', ylim=(-15, 125))
    ax.axhline(50, color='lightgray', lw=0.8, ls=':')


print('Plotting fitted curves...')
fig, axes = plt.subplots(1, 5, figsize=(17, 4.5), sharey=True)
colors = ['#0D9488', '#1B2A4A', '#E57373', '#F59E0B', '#6B7280']

for ax, (name, df), color in zip(axes, data.items(), colors):
    plot_one_curve(ax, name, df, fit_results[name], color)

axes[0].set_ylabel('Viability (%)')
fig.suptitle('4PL Curve Fitting Results — Five Drug Screening Scenarios\n'
             'Note: fitting always returns a number, but not all fits are reliable!',
             fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()

print('\nKey insight: Compound E (inactive) ALSO returns an IC50 value!')
print('The 4PL model will fit ANY data — a numerical IC50 does NOT mean the compound is active.')
print('This is why visual QC and quality metrics are essential.')
```

## Cell 8: Summary quality metrics table

**R-squared (R²):** How well the 4PL model explains the variance in the data.
- R² > 0.98: excellent fit
- R² 0.90–0.98: acceptable
- R² < 0.90: poor fit (high noise, wrong model, hook effect, etc.)

**RMSE:** Average prediction error in % viability units.
RMSE < 5% is good for a cell viability assay.

**CI width on IC50:** Narrow CI = confident IC50 estimate.
Wide CI = IC50 may be outside the tested range (incomplete curve).

```python
# ============================================================
# CELL 8: Quality metrics summary table
# ============================================================

# True IC50 values from the data generator (ground truth)
TRUE_IC50 = {
    'Compound A (clean)':      1.0,
    'Compound B (noisy)':      1.0,
    'Compound C (hook)':       0.5,    # meaningful only below the hook concentration
    'Compound D (incomplete)': 50.0,   # well outside tested range
    'Compound E (inactive)':   None,   # no IC50 (inactive)
}

print('=== Quality Metrics Summary ===')
print(f'{"Compound":<28} {"TrueIC50":>10} {"FitIC50":>10} {"CI_lo":>10} {"CI_hi":>10} {"R2":>8} {"RMSE":>8}')
print('-' * 100)

rows = []
for name, fit in fit_results.items():
    true_ic50 = TRUE_IC50[name]
    if fit['success']:
        ic50     = fit['ic50']
        ci_lo, ci_hi = fit['ic50_ci']
        ci_width = ci_hi - ci_lo
        r2       = fit['r_squared']
        rmse     = fit['rmse']
        pct_err  = abs(ic50 - true_ic50) / true_ic50 * 100 if true_ic50 else None
        print(f'{name:<28} {str(true_ic50):>10} {ic50:>10.3f} {ci_lo:>10.3f} {ci_hi:>10.3f} {r2:>8.3f} {rmse:>8.2f}%')
        rows.append({
            'Compound':          name,
            'True IC50 (µM)':    true_ic50,
            'Fitted IC50 (µM)':  round(ic50, 3),
            '95% CI lower':      round(ci_lo, 3),
            '95% CI upper':      round(ci_hi, 3),
            'CI width':          round(ci_width, 3),
            'R-squared':         round(r2, 3),
            'RMSE (%)':          round(rmse, 2),
            '% error vs true':   round(pct_err, 1) if pct_err is not None else 'N/A',
        })
    else:
        print(f'{name:<28} {str(true_ic50):>10} {"FAILED":>10}')

summary_df = pd.DataFrame(rows)
print('\n')
print(summary_df.to_string(index=False))

print('\n=== Interpretation ===')
print('Compound A: Low RMSE + R²~1 + narrow CI → highly reliable IC50')
print('Compound B: Higher RMSE + wider CI → IC50 is correct but less precise')
print('Compound C: Good R² but IC50 is WRONG due to hook effect (R² alone is not enough!)')
print('Compound D: Wide CI → IC50 is outside tested range, extrapolated value')
print('Compound E: IC50 value exists but compound is INACTIVE — R² near 0 is the tell')
```

## Cell 9: LLM-based visual curve quality assessment

An experienced scientist doesn't just look at numbers — they look at the curve shape.
LLMs with vision can do the same: identify hook effects, incomplete curves, and
other quality issues that numerical metrics miss.

We render each curve as a PNG image and ask Claude Sonnet to assess:
- Overall curve quality (Good / Acceptable / Poor)
- Whether the IC50 is reliable
- Any issues detected (hook, incomplete, noisy, etc.)
- Recommendation (Accept / Flag for repeat / Reject)

**This requires an Anthropic API key.** The cell is skipped if no key is found.

```python
# ============================================================
# CELL 9: LLM visual quality assessment (Claude Sonnet)
# ============================================================

if not ANTHROPIC_API_KEY:
    print('No Anthropic API key found — skipping LLM assessment.')
    print('Set ANTHROPIC_API_KEY in your .env file to enable this section.')
else:
    import anthropic

    # ---- Helper: render one curve to base64 PNG ----
    def render_curve_png(name, df, fit, color):
        """Render a curve image in memory and return base64 string for the LLM."""
        fig, ax = plt.subplots(figsize=(5, 4))
        conc = df['concentration_uM'].values
        resp = df['response_pct'].values

        ax.semilogx(conc, resp, 'o', color=color, ms=8, alpha=0.85)

        if fit['success']:
            conc_fine = np.logspace(np.log10(conc.min()) - 0.3,
                                    np.log10(conc.max()) + 0.3, 300)
            ax.semilogx(conc_fine, four_pl(conc_fine, *fit['params']),
                        '-', color=color, lw=2.5)
            ax.set_title(f"{name}  |  IC50={fit['ic50']:.2f}µM  R²={fit['r_squared']:.2f}",
                         fontsize=10)
        else:
            ax.set_title(f"{name}  |  fit failed", fontsize=10)

        ax.set(xlabel='Concentration (µM)', ylabel='Viability (%)', ylim=(-15, 125))
        ax.axhline(50, color='lightgray', lw=0.8, ls=':')
        plt.tight_layout()

        buf = io.BytesIO()
        fig.savefig(buf, format='png', dpi=150, bbox_inches='tight')
        plt.close(fig)
        buf.seek(0)
        return base64.standard_b64encode(buf.read()).decode('utf-8')

    # ---- Helper: call Claude Sonnet for assessment ----
    def assess_with_claude(name, df, fit, color, client):
        """Send curve image to Claude and return structured quality assessment."""
        img_b64 = render_curve_png(name, df, fit, color)

        prompt = """You are a medicinal chemist reviewing dose-response curves from a cell viability assay.

In a VIABILITY assay: y-axis = cell viability (%). A healthy curve starts near 100%
(low drug concentration = no effect) and falls toward 0% (high concentration = full kill).
The curve should slope DOWNWARD from left to right.

Assess this dose-response curve and respond in EXACTLY this format:

QUALITY: [Good / Acceptable / Poor]
IC50_RELIABLE: [Yes / No / Uncertain]
ISSUES: [comma-separated list of issues, or 'None']
REASONING: [2-3 sentences explaining your assessment]
RECOMMENDATION: [Accept / Flag for repeat / Reject]

Consider: sigmoid completeness, hook effect, noise level, plateau visibility,
and whether the IC50 falls within the tested concentration range."""

        response = client.messages.create(
            model='claude-sonnet-4-5',
            max_tokens=400,
            messages=[{
                'role': 'user',
                'content': [
                    {'type': 'image',
                     'source': {'type': 'base64', 'media_type': 'image/png',
                                'data': img_b64}},
                    {'type': 'text', 'text': prompt},
                ]
            }]
        )
        return response.content[0].text

    # ---- Run assessments ----
    print('Sending each curve to Claude Sonnet for quality assessment...')
    print('(This makes API calls — each takes ~2–5 seconds)')
    print()

    client = anthropic.Anthropic(api_key=ANTHROPIC_API_KEY)
    colors = ['#0D9488', '#1B2A4A', '#E57373', '#F59E0B', '#6B7280']
    llm_assessments = {}

    for (name, df), color in zip(data.items(), colors):
        print(f'Assessing: {name}')
        assessment = assess_with_claude(name, df, fit_results[name], color, client)
        llm_assessments[name] = assessment
        print(assessment)
        print('-' * 60)

    print('\nAll LLM assessments complete!')
```

```python
# ============================================================
# CELL 10: Final summary visualisation (curve + LLM verdict)
# ============================================================

def parse_field(text, field):
    """Extract a single field from the structured LLM response."""
    for line in text.split('\n'):
        if line.startswith(field + ':'):
            return line.split(':', 1)[1].strip()
    return 'N/A'


verdict_colors = {'Good': '#0D9488', 'Acceptable': '#F59E0B', 'Poor': '#E57373'}

if 'llm_assessments' not in dir():
    print('LLM assessments not available (no API key). Showing curve plots only.')
    fig, axes = plt.subplots(1, 5, figsize=(17, 4.5), sharey=True)
    colors = ['#0D9488', '#1B2A4A', '#E57373', '#F59E0B', '#6B7280']
    for ax, (name, df), color in zip(axes, data.items(), colors):
        plot_one_curve(ax, name, df, fit_results[name], color)
    axes[0].set_ylabel('Viability (%)')
    fig.suptitle('4PL Curve Fitting — Five Scenarios', fontsize=12, fontweight='bold')
    plt.tight_layout()
    plt.show()
else:
    print('Creating combined curve + LLM verdict figure...')
    fig, axes = plt.subplots(2, 5, figsize=(17, 7),
                             gridspec_kw={'height_ratios': [3, 1]})
    colors = ['#0D9488', '#1B2A4A', '#E57373', '#F59E0B', '#6B7280']

    for col, (name, df), color in zip(range(5), data.items(), colors):
        # Row 0: curve
        plot_one_curve(axes[0][col], name, df, fit_results[name], color)
        # Row 1: LLM verdict text box
        ax_text = axes[1][col]
        ax_text.axis('off')
        assessment = llm_assessments.get(name, '')
        quality = parse_field(assessment, 'QUALITY')
        rec     = parse_field(assessment, 'RECOMMENDATION')
        issues  = parse_field(assessment, 'ISSUES')[:45]
        vc      = verdict_colors.get(quality, '#888')
        ax_text.text(0.5, 0.5,
                     f'Quality: {quality}\n{rec}\n{issues}',
                     ha='center', va='center', fontsize=8,
                     color=vc, fontweight='bold',
                     transform=ax_text.transAxes,
                     bbox=dict(boxstyle='round,pad=0.3',
                               facecolor=vc+'22', edgecolor=vc, lw=1.2))

    fig.suptitle('Dose-Response Analysis: Fitted Curves + LLM Quality Assessment',
                 fontsize=12, fontweight='bold')
    plt.tight_layout()
    plt.show()
    print('Combined figure generated.')

print('\n=== Final Summary ===')
print('Numerical metrics (R², RMSE, CI) catch SOME problems but not all.')
print('Visual inspection (by human or LLM) catches hook effects and contextual issues.')
print('Best practice: use BOTH together for robust IC50 quality control.')
```
